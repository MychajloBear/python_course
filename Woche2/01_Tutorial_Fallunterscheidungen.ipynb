# 6) `If-Else` Bedingung

`If-Else` Bedingungen werden verwendet, wenn Code abhängig von einer (oder mehreren) Bedingung ausgeführt werden soll. 

So kann z.B. Code nur dann ausgeführt werden, wenn eine Variable einen bestimmten Wert hat oder ein bestimmter Buchstabe eingegeben wurde. 

Ohne `If-Else` Bedingungen wären unsere Programme "linear", d.h. der Programmfluss wäre klar und rein deterministisch. 

Mit `If-Else` Bedingungen können wir unseren Code dynamischer und flexibler gestalten.
### 6.1) Syntax:

Die Syntax für `If-Else` Bedingungen funktioniert in Python über **Einr�ckung des Codes**. 

Dafür wird normalerweise ein Tabulatorzeichen (oder 4 Leerzeichen) verwendet (siehe Beispiele). 

Der eingerückte Code gehört dann zum Geltungsbereich (**scope**) der `If-Else` Bedingung.

```python
if Bedingung:
    Code der nur ausgeführt wird, wenn die Bedingung True ist
```
`Bedingung` ist hierbei ein Ausdruck, der entweder `True` oder `False` ist.




 **Beispiele**

Im folgenden werden zwei `print()`-Befehle ausgegeben. Die Bedingung `True` ist trivialerweise immer erfüllt:

````python
if True:
    print("Die Bedingung ist erfüllt.")
    print("Der eingerückte Code wird ausgeführt, wenn die Bedigung erfüllt ist.")
````

Diese Bedingung `True` ist immer erfüllt (wahr). Wir bräuchten in diesem Fall keine `If`-Bedingung.

Wir können auch eine `If`-Bedingung mit einer Bedingung schreiben, die niemals erfüllt ist, d.h. der eingerückte Code wird niemals ausgeführt:

````python
if False:
    print("Diese Zeile wird nie ausgeführt, Bedingung ist False")

print("Diese Zeile wird ausgeführt und gehört NCIHT zur If-Bedingung.")
````

### 6.2) Mehrere Bedingungen überprüfen (`elif`)

Wollen wir mehr als eine Bedingung überprüfen, nutzen wir folgende Syntax:

```python
if Bedingung1:
    Code der nur ausgeführt wird, wenn Bedingung1 True ist
elif Bedingung2:
    Code der nur ausgeführt wird, wenn Bedingung2 True ist
```

Wenn `Bedingung1` **nicht** erfüllt, prüft Python ob `Bedingung2` erfüllt ist, und führt gegebenenfalls den eingerückten Code aus. 

Wir können beliebige viele `elif` Bedingungen schreiben. 

Ist jedoch eine Bedingung erfüllt, werden **keine** weiteren Bedingungen überprüft, sondern das If-Konstrukt verlassen. 

Das bedeutet, dass die **Reihenfolge** der Bedingungen von Bedeutung ist.

Ein Gleichheitszeichen `=` dient der Wertzuweisung.\
Zwei Gleichheitszeichen `==` prüfen die Gleichheit zweier Werte (boolesche Gleichung).

Im folgenden Programm sind beide Bedingungen erfüllt, dennoch wird **nur** die erste `elif` Anweisung ausgeführt. 


````python
z = 5

if z > 6:     # Nicht erfüllt.
    print("z > 6")

elif z < 6:   # Erfüllt
    print("z < 6")

elif z == 5:  # Erfüllt
    print("z = 5")     #Wird nicht ausgegeben, da schon eine Bedingung vorher erfüllt wurde.
    
print("Hier sind wir wieder im Hauptprogramm.")
````


### 6.2) Mehrere Bedingungen überprüfen mit (`elif`) und (`else`)
Wenn du mehr als nur zwei Möglichkeiten überprüfen musst, kannst du deiner ``if``-Abfrage mit ``elif`` (kurz für ``else if``) um beliebig viele Bedingungen erweitern und mit ``else`` einen Standardfall festlegen.

### Syntax:

```python
if Bedingung1:
    Code der nur ausgeführt wird, wenn Bedingung1 True ist
elif Bedingung2:
    Code der nur ausgeführt wird, wenn Bedingung2 True ist
else: 
    Code der nur ausgeführt wird, wenn keine Bedingung vorher erfüllt wurde
```

Der Code innerhalb des `else`-Blocks wird ausgeführt, wenn keine der vorherigen Bedingungen erfüllt wurde.

#### Beispiel:
 
 Im folgenden Beispiel lohnt es sich mit unterschiedlichen Werten von `a` zu experimentieren. 

````python
a = 3

if a > 3:   
    print("a > 3")

elif a == 5:
    print("a == 5")
    
else:
    print("Keine Bedingung wurde erfüllt.")
````

### 6.5) Best Practice: Verschachtelung vs. Logische Operatoren

Manchmal sieht man Code, bei den ``if``-Abfragen tief ineinander verschachtelt werden, um einen Bereich zu prüfen. Das funktioniert zwar theoretisch, sorgt aber schnell für unübersichtlichen Code.

**Negativ-Beispiel: Unübersichtlich durch Verschachtelung**

````python
y = 5

if y > 10:
    if y < 100:
        print("y ist zwischen 10 und 100.")
    
else:
    print(y, "ist nicht zwischen 10 und 100.")
````

**Positiv-Beispiel: Elegant und lesbar mit ``and``**

````python
# besser:
y = 5

if y > 10 and y < 100:
    print(y, "ist zwischen 10 und 100.")
    
else:
    print(y, "ist nicht zwischen 10 und 100.")
````
