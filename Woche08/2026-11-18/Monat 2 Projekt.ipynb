# Einfluss der Futterart auf das Gewicht von Hühnern

In diesem Projekt sollst du untersuchen, ob die Art des Futters einen Einfluss auf das Gewicht von Hühnern hat. Dazu werden Gewichtsdaten von Hühnern analysiert, die mit unterschiedlichen Futtersorten gefüttert wurden. Ziel ist es, Zusammenhänge zwischen der Fütterung und der Gewichtsentwicklung der Tiere zu erkennen und mithilfe statistischer Verfahren zu überprüfen.


## Datensatz
Der zugrunde liegende Datensatz stammt aus einem Fütterungsversuch mit 50 Hühnern. Jedes Huhn wurde einer von vier Futterarten zugeordnet und über einen Zeitraum von 21 Tagen beobachtet. Während des Versuchs wurde das Gewicht der Tiere in regelmäßigen Abständen gemessen und dokumentiert. Neben dem Gewicht enthält der Datensatz Informationen zum Alter des Huhns, zur individuellen Huhn-ID sowie zur verwendeten Futterart.

| Variable | Datentyp | Skalenniveau | Beschreibung |
|-----------|-----------|-----------|-----------|
| `huhn` | Integer | Nominal | Eindeutige Identifikationsnummer eines Huhns |
| `alter` | Integer | Metrisch (Verhältnisskala) | Alter des Huhns zum Zeitpunkt der Messung in Tagen |
| `gewicht` | Integer | Metrisch (Verhältnisskala) | Gewicht des Huhns zum jeweiligen Messzeitpunkt in Gramm |
| `futter` | Integer | Nominal | Zugeordnete Futterart des Huhns |

## Ziel der Analyse
Im Rahmen der Analyse werden zunächst die Gewichte der Hühner über die Zeit beschrieben und visualisiert. Anschließend wird untersucht, ob sich die Gewichte zwischen den verschiedenen Futtergruppen unterscheiden und ob diese Unterschiede statistisch signifikant sind. Dadurch kann beurteilt werden, welchen Einfluss die Futterart auf das Wachstum der Hühner hat.
