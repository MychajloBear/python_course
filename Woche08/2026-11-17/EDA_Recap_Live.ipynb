# 1. EDA-Recap: Der EDA Prozess
Wir haben in den letzten drei Wochen vieles gelernt: Die Grundlagen von Numpy und Pandas, den Prozess der Datensatz-Bereinigung und -Transformation sowie den idealen Einsatz von Visualisierungen, um unsere Analysen zu stützen oder darzustellen.

In der heutigen Unterrichtseinheit wollen wir noch einmal Schritt-für-Schritt über den EDA Prozess sprechen und die wichtigsten Konzepte gemeinsam wiederholen, damit ihr fit seid, mit fremden Datensätzen zielsicher umzugehen!

### Die drei Ziele der EDA
#### 1. Daten verstehen
Was für Variablen gibt es, was bedeuten sie fachlich?
#### 2. Datenqualität einschätzen
Wo gibt es Lücken, Fehler, Inkonsistenzen — und wie stark beeinträchtigen sie die Aussagekraft?
#### 3. Entscheidungsgrundlage schaffen
Für Bereinigung, Modellierung, oder eine fachliche Empfehlung.

### Die vier Phasen
* Datenbereinigung — fehlende Werte, Duplikate, unplausible Werte, inkonsistente Kategorien
* Datentransformation — Typen korrigieren, neue Merkmale ableiten, kodieren
* Visualisierung — einzelne Variablen und Zusammenhänge sichtbar machen
* Interpretation — Muster fachlich einordnen, nächste Schritte ableiten

**Der wichtige Punkt: Es ist ein Kreislauf, keine Linie** <br>
In der Praxis springt man ständig zwischen diesen Phasen. Eine Visualisierung deckt zum Beispiel oft erst auf, dass noch etwas bereinigt werden muss (z.B. ein sichtbarer Ausreißer im Histogramm). Eine neue transformierte Variable wirft manchmal eine neue Frage auf, die wieder eine Bereinigung nötig macht.
* Typischer Ablauf in der Realität: erste grobe Sichtung → grobe Bereinigung → erste Visualisierung → dadurch neue Auffälligkeiten entdeckt → gezieltere Bereinigung → Transformation → weitere Visualisierung → Interpretation → evtl. zurück zum Anfang mit neuer Fragestellung.
* EDA hat deshalb kein festes Ende im klassischen Sinn — man hört auf, wenn die Fragen, die man beantworten wollte, hinreichend beantwortet sind (oder das Zeitbudget aufgebraucht ist).

### Entscheidungsleitende Fragen
#### Vor jeder Entscheidung: Kontext klären
Die meisten EDA-Fehler entstehen nicht durch falsche Syntax, sondern durch Entscheidungen ohne Kontext. Bevor irgendetwas bereinigt oder entfernt wird, sollten diese Fragen beantwortet sein:
* Woher stammen die Daten? Wer hat sie erhoben, mit welchem Prozess?
* Was bedeutet ein fehlender Wert fachlich? (Wurde nichts erfasst? Gab es das Merkmal nicht? Ein Systemfehler?)
* Welche Frage soll der Datensatz überhaupt beantworten? Ohne Zielfrage lässt sich "gut" oder "schlecht bereinigt" gar nicht beurteilen.
#### Framework: Fehlende Werte einordnen (konzeptuell)
* Zufällig fehlend, unabhängig von allen Variablen: meist unproblematisch, einfache Verfahren reichen.
* Zufällig fehlend, aber abhängig von einer bekannten anderen Variable (z.B. ältere Verträge haben öfter keine E-Mail-Adresse): erfordert   differenziertere Behandlung, ggf. je Gruppe.
* Nicht zufällig fehlend — das Fehlen selbst trägt Information (z.B. Kund:innen mit niedrigem Einkommen geben es seltener an): hier ist Löschen oder naives Auffüllen besonders riskant, da es das Muster verzerrt.
#### Framework: Auffällige/extreme Werte einordnen
* Ist der Wert technisch möglich, aber fachlich unwahrscheinlich? → Rücksprache mit Fachbereich statt automatischem Löschen.
* Ist der Wert ein offensichtlicher Erfassungsfehler (z.B. negatives Alter)? → Korrektur oder Entfernen meist gerechtfertigt.
* Ist der Wert ein echtes, seltenes Ereignis (z.B. Umsatzspitze am Black Friday)? → Entfernen würde echte Information vernichten.

Deshalb: „Ausreißer entfernen“ ist niemals ein Automatismus, sondern immer eine fachliche Abwägung — unabhängig davon, mit welcher Methode man den Wert überhaupt gefunden hat.
#### Die Zielfrage als Kompass
Jede Bereinigungs-, Transformations- und Visualisierungsentscheidung sollte man an der Ausgangsfrage spiegeln: Hilft mir dieser Schritt, die eigentliche Frage besser zu beantworten — oder bereinige ich, weil es "sauberer aussieht"?

<div class="alert alert-block alert-info">
<b>Merksatz</b> Eine Entscheidung ohne nachvollziehbare Begründung ist keine EDA-Entscheidung, sondern eine Vermutung. Jede Bereinigung sollte in einem Satz begründbar sein.</div>

### Häufige Fallstricke und Denkfehler
**1. Korrelation ≠ Kausalität** <br>
Zwei Variablen können stark zusammenhängen, ohne dass eine die andere verursacht — oft steckt eine dritte, gemeinsame Ursache dahinter (Konfundierung), oder der Zusammenhang ist schlicht Zufall.
* Klassisches Beispiel: Eisverkauf und Ertrinkungsfälle korrelieren — beide hängen tatsächlich von der Jahreszeit/Temperatur ab.

**2. Cherry-Picking / Bestätigungsfehler**<br>
Man sucht unbewusst nach Mustern, die die eigene Vorannahme bestätigen, und übersieht Gegenbeispiele. Gegenmittel: bewusst nach Daten suchen, die die eigene Hypothese widerlegen könnten.

**3. Data Dredging / multiples Testen**<br>
Wer genug Variablenkombinationen durchprobiert, findet garantiert irgendwo einen "auffälligen" Zusammenhang — rein durch Zufall. Das ist bei explorativer Analyse besonders verführerisch, weil man ja gezielt nach Mustern sucht. (Zitat Ronald Coase: if you torture the data long enough, it will confess to anything)
* Gegenmittel: gefundene Muster als Hypothese behandeln, die noch bestätigt werden muss — nicht als bereits bewiesene Erkenntnis.

**4. Blindes Vertrauen in Standardverfahren**<br>
Eine Methode (z.B. eine bestimmte Imputationsstrategie oder Visualisierungsart) "weil man sie halt kennt" einzusetzen, ohne zu prüfen, ob sie zum Datentyp und zur Fragestellung passt, führt regelmäßig zu falschen Schlussfolgerungen.

**5. Data Leakage (Informationslecks)**<br>
Wenn bei der Exploration oder späteren Modellierung versehentlich Informationen einfließen, die zum Zeitpunkt der eigentlichen Fragestellung noch gar nicht bekannt gewesen wären (z.B. ein Merkmal, das erst nach dem Zielereignis entsteht), wirkt die Analyse besser, als sie in der Realität wäre.

**6. Stichprobenverzerrung (Selection Bias)**<br>
Wenn der vorliegende Datensatz selbst schon eine verzerrte Auswahl ist (z.B. nur zufriedene Kund:innen antworten auf eine Umfrage), führt jede noch so sorgfältige EDA zu einem verzerrten Bild der Grundgesamtheit.

**Reflexionsfrage:** Welchen der 6 Fallstricke haltet ihr für am schwersten zu erkennen?

### Ergebnisse korrekt kommunizieren
**EDA ist nicht nur für einen selbst**<br>
Eine EDA, die nur im eigenen Kopf oder in unkommentierten Code-Zellen existiert, ist für andere nicht nachvollziehbar und geht mit der Zeit auch für einen selbst verloren. Eine gute EDA-Zusammenfassung macht Entscheidungen und Grenzen transparent.

**Was in eine gute Zusammenfassung gehört**
* Getroffene Annahmen — was wurde vorausgesetzt, wo bestand Interpretationsspielraum?
* Getroffene Entscheidungen mit Begründung — z.B. warum wurden bestimmte Zeilen entfernt?
* Bekannte Einschränkungen — was bleibt unsicher, was konnte nicht geklärt werden?
* Offene Fragen — was sollte als Nächstes untersucht werden?

**Zielgruppe mitdenken**<br>
Für Fachbereich/Stakeholder zählt vor allem die fachliche Einordnung ("Was bedeutet das für uns?"), für Kolleg:innen im Data-Team eher die methodische Nachvollziehbarkeit ("Wie kam die Zahl zustande?"). Beides sollte im Idealfall verfügbar, aber nicht vermischt sein.

### Abschluss: Kurze "Trockenübung"
**Szenario 1 — Kundenabwanderung**<br>
In einem Datensatz zur Kundenabwanderung fehlen 40 % der Werte in der Spalte „Vertragsart“. Was fragt ihr zuerst, bevor ihr euch für eine Bereinigungsstrategie entscheidet?<br>
**Szenario 2 — Umsatz-Ausreißer**<br>
In Verkaufsdaten seht ihr einen extremen Umsatz-Ausreißer am Black-Friday-Wochenende. Entfernen oder behalten? Welche Argumente sprechen für welche Seite?<br>
**Szenario 3 — Verdächtige Korrelation**<br>
Zwei Variablen in eurem Datensatz korrelieren auffällig stark. Welche drei Fragen stellt ihr euch, bevor ihr daraus einen inhaltlichen Schluss zieht?<br>
**Szenario 4 — Auffälliges Muster bei Bewerbungsdaten**<br>
Bei der Analyse von Bewerbungsdaten fällt auf: Bewerber:innen einer bestimmten Universität werden deutlich häufiger eingeladen. Ist das ein Datenproblem, ein reales (evtl. problematisches) Muster, oder beides? Wie geht ihr weiter vor?<br>
**Szenario 5 — Zeitdruck vs. Datenqualität**<br>
Der Product Owner braucht „schnell mal eine Einschätzung“, aber die Datenqualität ist fraglich, es gibt große Lücken. Wie kommuniziert ihr das, ohne entweder ganz abzulehnen oder eine falsche Sicherheit vorzutäuschen?<br>
**Szenario 6 — Viele Tests, wenige Treffer**<br>
Ihr habt 20 verschiedene Variablenkombinationen auf einen Zusammenhang mit der Zielgröße geprüft und dabei 2 „signifikante“ gefunden. Wie ordnet ihr dieses Ergebnis ein?

## Datenbereinigung
