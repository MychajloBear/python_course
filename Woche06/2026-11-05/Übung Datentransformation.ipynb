## Szenario
Die FrogTech Solutions GmbH ist ein mittelständisches Technologieunternehmen mit 90 Mitarbeitenden in Deutschland. Die Mitarbeitenden arbeiten an den Standorten Berlin, München, Hamburg und Köln und sind auf die Abteilungen IT, HR, Finance, Sales und Marketing verteilt.

Die Personalabteilung möchte ihre Daten künftig stärker datengetrieben nutzen und hat deshalb einen Datensatz mit Informationen zu allen Mitarbeitenden zusammengestellt.

Der Datensatz wurde bereits bereinigt. Die Namen liegen in einem einheitlichen Format vor und die Informationen sind vollständig und korrekt. Allerdings möchte die Personalabteilung noch einige Informationen anders darstellen und neue Spalten hinzufügen, um später sinnvolle Analysen damit machen zu können. Hilf der Personalabteilung dabei, das umzusetzen.

Verschaffe dir einen Überblick über die Daten der Mitarbeitenden.

    * Achte dabei darauf, dass die erste Spalte (die 0. Spalte) den Index enthält.
    * Achte darauf, die Spalte `hire_date` mit `parse_dates` als Datumsspalte einzulesen.
2. Lass dir die Anzahl an Zeilen und Spalten, sowie die Spaltennamen und deren Datentypen anzeigen.
3. Gib anschließend die ersten fünf Zeilen aus.

## Aufgabe 2 – Geschlecht in deutsche Kategorien umwandeln

Die Personalabteilung möchte die Geschlechtsangaben im Datensatz verständlicher darstellen.

Aktuell enthält die Spalte gender die folgenden Kürzel: 'd', 'f', 'm'.

Erstelle eine neue Spalte namens `gender_GER`, in der die Geschlechter auf Deutsch ausgeschrieben werden.

Verwende dabei folgende Zuordnung:
|Kürzel|Bezeichnung|
|--|--|
|d|Divers|
|f|Weiblich|
|m|Männlich|

<details>
  <summary> Hinweis anzeigen</summary>

* Verwende ein Python-Dictionary für die Zuordnung der Werte.
* Nutze die Methode `.map()`, um die Werte der Spalte `gender` zu übersetzen.
* Speichere das Ergebnis in einer neuen Spalte namens `gender_GER`.
* Die ursprüngliche Spalte `gender` soll erhalten bleiben.

</details>

```python
gender_mapping = {
    "m": "Männlich",
    "f": "Weiblich",
    "d": "Divers"
}

df["gender_GER"] = df["gender"].map(gender_mapping)

df[["gender", "gender_GER"]].head()
```

Aktuell enthält die Spalte `department` folgende Werte:

Ersetze die beiden Abkürzungen durch ausgeschriebene Werte. Die übrigen Werte sollen unverändert bleiben.

| Kürzel | Bezeichnung            |

Speichere das Ergebnis in einer neuen Spalte `department_report`.

## Aufgabe 4 – Beschäftigungsdauer

Die Personalabteilung möchte wissen, wie lange die Mitarbeitenden bereits im Unternehmen beschäftigt sind.

Erstelle eine neue Spalte namens `years_at_company`, die die Anzahl der Jahre seit dem Einstellungsdatum enthält. Gehe dabei wie folgt vor:

1. Ermittle das heutige Datum.
2. Berechne die Zeitspanne zwischen dem heutigen Datum und dem Einstellungsdatum (`hire_date`).
3. Wandle die Zeitspanne in Jahre um.
4. Speichere das Ergebnis in der neuen Spalte `years_at_company`.
5. Runde das Ergebnis auf eine Nachkommastelle.

<details>
  <summary> Hinweis anzeigen</summary>

* Nutze `pd.Timestamp.today()`, um das heutige Datum zu ermitteln.
* Wandle die Spalte `hire_date` mit `pd.to_datetime()` in ein Datumsformat um.
* Berechne die Differenz zwischen dem heutigen Datum und dem Einstellungsdatum.
* Die Anzahl der Tage kannst du mit `.dt.days` aus der Zeitdifferenz extrahieren.
* Teile die Anzahl der Tage durch `365.25`, um die Beschäftigungsdauer in Jahren zu berechnen.
* Runde das Ergebnis mit `.round(1)` auf eine Nachkommastelle.

</details>



```python
today = pd.Timestamp.today()

df["years_at_company"] = (
    (today - df["hire_date"]).dt.days
    / 365.25
).round(1)

df[["hire_date", "years_at_company"]].head()
```

## Aufgabe 5 – Verteilung der Abteilungen

Die Personalabteilung möchte einen Überblick darüber erhalten, wie viele Mitarbeitende in den einzelnen Abteilungen beschäftigt sind.

Ermittle die Anzahl der Mitarbeitenden pro Abteilung und stelle das Ergebnis übersichtlich dar.

Gehe dabei wie folgt vor:

1. Gruppiere die Daten nach der Spalte `department`.
2. Zähle die Anzahl der Mitarbeitenden in jeder Gruppe.

<details>
<summary>Hinweis anzeigen</summary>

## Aufgabe 6 – Verteilung der Abteilungen und Geschlechter

Die Personalabteilung möchte prüfen, wie sich die Mitarbeitenden verschiedener Geschlechter auf die einzelnen Abteilungen verteilen.

Erstelle eine Kreuztabelle, die für jede Abteilung zeigt, wie viele Mitarbeitende den Geschlechtergruppen männlich (m), weiblich (f) und divers (d) angehören.

Gehe dabei wie folgt vor:

1. Verwende die Spalten `department` und `gender`.
2. Erstelle eine Kreuztabelle, die die Anzahlen an Mitarbeitenden in den Kombinationen der Kategorien anzeigt.
3. Interpretiere das Ergebnis:
    * In welcher Abteilung arbeiten die meisten Mitarbeitenden?
    * Welche Abteilungen haben die höchste Anzahl weiblicher Mitarbeitender?
    * In welchen Abteilungen arbeiten diverse Mitarbeitende?


<details>
<summary>Hinweis anzeigen</summary>

* Nutze `pd.crosstab()`, um zwei kategoriale Variablen miteinander zu vergleichen.
* Verwende für eine gute Übersicht die Spalte `department` als Zeilen und `gender` als Spalten.
* Die Werte in der Tabelle zeigen die Anzahl der Mitarbeitenden pro Kombination.

</details>

```python
gender_department = pd.crosstab(
    df["department"],
    df["gender"]
)

print(gender_department)

gender_department.sum(axis=1).idxmax()
gender_department["f"].idxmax()
gender_department[gender_department["d"] > 0]
```

## Aufgabe 7 – Einteilung in Erfahrungsstufen

Die Personalabteilung möchte die Mitarbeitenden anhand ihrer Beschäftigungsdauer in verschiedene Erfahrungsstufen einteilen.

Erstelle eine neue Spalte namens `experience_level`, die jede Person einer der folgenden Kategorien zuordnet:

|Beschäftigungsdauer|	Kategorie|
|----|----|
|Weniger als 3 Jahre|	Junior|
|3 bis unter 7 Jahre|	Medior|
|7 Jahre oder mehr|	Senior|

Gehe dabei wie folgt vor:

1. Nutze dafür `pd.cut()` für die Spalte `years_at_company`.
2. Speichere das Ergebnis in einer neuen Spalte `experience_level`.

<details>
<summary>Hinweis anzeigen</summary>

* Nutze die Liste `[0, 3, 7, float("inf")]` für die `bins`.
* Nutze die Liste `["Junior", "Medior", "Senior"]` für die `labels`.

</details>


Alternative Lösung:
1. Schreibe eine Funktion namens `classify_experience`, die die Beschäftigungsjahre entgegennimmt und die Erfahrungsstufe ausgibt.
2. Wende diese Funktion auf die Spalte `years_at_company` an.
3. Speichere das Ergebnis in einer neuen Spalte `experience_level`.

<details>
<summary>Hinweis anzeigen</summary>

* Nutze `apply()`, um für jede Zeile eine Kategorie zu bestimmen.
* Verwende eine Funktion mit `if`, `elif` und `else`.

</details>

```python
df["experience_level"] = pd.cut(
    df["years_at_company"],
    bins=[0, 3, 7, float("inf")],
    labels=["Junior", "Medior", "Senior"],
    right=False
)
```

```python
def classify_experience(years):
    if years < 3:
        return "Junior"
    elif years < 7:
        return "Medior"
    else:
        return "Senior"


df["experience_level"] = (
    df["years_at_company"]
    .apply(classify_experience)
)
```

## Aufgabe 8 – Gehaltsunterschiede

Die Personalabteilung möchte untersuchen, wie sich die Gehälter innerhalb der verschiedenen Abteilungen und Erfahrungsstufen unterscheiden.

Erstelle eine Pivot-Tabelle, die das durchschnittliche Gehalt für jede Kombination aus Abteilung (`department`) und Beschäftigungsgruppe (`experience_level`) anzeigt.

<details>

<summary>Hinweis anzeigen</summary>

* Verwende `pd.pivot_table()`.
* Nutze `department` als `index`.
* Nutze `experience_level` als `columns`.
* Verwende `salary` als `values`.
* Berechne den Durchschnitt mit `aggfunc="mean"`.

</details>


```python
salary_pivot = pd.pivot_table(
    df,
    values="salary",
    index="department",
    columns="experience_level",
    aggfunc="mean"
)

print(salary_pivot)
```

## Aufgabe 9 - Bonusberechtigte Mitarbeitende identifizieren

Die Personalabteilung möchte besonders leistungsstarke Mitarbeitende finanziell fördern. Allerdings soll der Bonus nur an Mitarbeitende vergeben werden, die eine sehr gute Leistungsbewertung haben und gleichzeitig noch nicht zu den Spitzenverdienenden gehören.

Erstelle eine neue Spalte namens `bonus_eligible`.

Verwende dabei folgende Regeln:

* Performance Score = 5 UND Gehalt < 80.000 €
→ Ja
* Sonst
→ Nein

Gehe dabei wie folgt vor:

1. Schreibe eine Funktion namens `check_bonus()`.
    * Prüfe für jede Person den `performance_score`.
    * Prüfe zusätzlich das `salary`.
    * Weise anhand der beiden Bedingungen die Werte "Ja" oder "Nein" zu.
2. Wende die Funktion auf den DataFrame `df` an.
3. Speichere das Ergebnis in einer neuen Spalte `bonus_eligible`.

<details>
<summary>Hinweis anzeigen</summary>

* Die Funktion `check_bonus()` soll eine Zeile übergeben bekommen und als Rückgabewert entweder "Ja" oder "Nein" zurückgeben.
* Die Überprüfung der Leistungsbewertung sieht so aus: `row["performance_score"] == 5`.
* Schreibe eine entsprechende Überprüfung des Gehalts und verknüpfe beide Überprüfungen mit einer UND-Verknüpfung.
* Verwende `if` und `else` in der Funktion `check_bonus()`.
* Nutze `apply()` mit `axis=1`.

</details>

```python
def check_bonus(row):
    if (
        row["performance_score"] == 5
        and row["salary"] < 80000
    ):
        return "Ja"

    return "Nein"


df["bonus_eligible"] = (
    df.apply(check_bonus, axis=1)
)

df[["performance_score", "salary", "bonus_eligible"]].head()
```

## Aufgabe 10 - Abteilung mit höchster durchschnittlicher Leistungsbewertung

Die Personalabteilung möchte herausfinden, welche Abteilung im Durchschnitt die besten Leistungsbewertungen erzielt.

Berechne die durchschnittliche Leistungsbewertung für jede Abteilung und ermittle anschließend die Abteilung mit dem höchsten Durchschnittswert.

Gehe dabei wie folgt vor:

1. Gruppiere die Daten nach `department`.
2. Berechne den durchschnittlichen `performance_score` pro Abteilung.
3. Sortiere das Ergebnis absteigend.
4. Bestimme die Abteilung mit der höchsten durchschnittlichen Leistungsbewertung.

<details>
<summary>Hinweis anzeigen</summary>

* Nutze `df.groupby()`, um die Daten nach Abteilung zu gruppieren.
* Verwende `.mean()`, um den durchschnittlichen Performance Score zu berechnen.
* Sortiere die Werte mit `.sort_values()`.
* Die erste Zeile enthält anschließend die Abteilung mit dem höchsten Durchschnittswert.

</details>

```python
performance_avg = (
    df.groupby("department")["performance_score"]
      .mean()
      .sort_values(ascending=False)
)

print(performance_avg)
```
