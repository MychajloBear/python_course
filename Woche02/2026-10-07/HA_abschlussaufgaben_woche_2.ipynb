## Aufgabe 6
### Tag 2
Berechne die Summe der Quadrate der ersten fünfzehn natürlichen Zahlen (von `1` bis `15`) mit einer `for`-Schleife.

```python
sum_of_squares = 0

for number in range(1, 16):
    sum_of_squares += number ** 2

print("Summe der Quad von 1 bis 15:", sum_of_squares)
```

## Aufgabe 9
### Tag 2

Verwende eine `while`-Schleife, um die größte Zahl der Liste `max_list` zu finden und das Ergebnis in `max_value` zu speichern. Vergleiche anschließend das Ergebnis mit der `max()`-Funktion.

```python
# Lasse den folgenden Code unverändert und führe nur die Zelle aus

max_list = [3, 7, 2, 9, 5, 12, 8, 5, 4, 1]
max_value = 0
```

```python

max_value = max_list[0]
index = 0

while index < len(max_list):
    if max_list[index] > max_value:
        max_value = max_list[index]
    index += 1

print("mit while-Schleife:", max_value)
print("mit max()-Funktion:", max(max_list))
```
