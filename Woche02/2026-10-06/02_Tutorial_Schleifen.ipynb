## 1) `for`-Schleifen

**`for`-Schleifen** dienen dazu, einen bestimmten Codeabschnitt *wiederholt* auszuführen. Dadurch lassen sich wiederkehrende Aufgaben automatisieren, ohne dass derselbe Code mehrfach geschrieben werden muss.

In Python werden `for`-Schleifen über eine Sequenz ausgeführt und für jedes Element dieser Sequenz wird der Code *innerhalb* der Schleife ausgeführt. Konkret lautet die Syntax:

```python
for element in sequence:
    Code, der für jedes Element der Sequenz ausgeführt wird
```

Hierbei ist 
- `sequence`: Eine Sequenz von Elementen, über die iteriert wird.
- `element`: In jedem Schleifendurchlauf wird `element` das aktuelle Element aus `sequence` zugewiesen.

> <span style="color:red"> **_HINWEIS:_** </span> Der auszuführende Code muss wie bei `if`-Bedingungen *eingerückt* werden.

**Beispiel:**

Das folgende Beispiel zeigt eine `for`-Schleife angewendet auf eine Liste `my_list`. 

Bei jedem **Schleifendurchlauf**, bzw. jeder **Iteration** wird das nächste Element aus `my_list` genommen und in der Variable `element` gespeichert. 

> <span style="color:red"> **_HINWEIS:_** </span> Statt `element` können wir für den sogenannten **Iterator** einen beliebigen Namen verwenden. Der Name dient somit nur als Platzhalter.

````python
my_list = ["A", 1, 2, "B"]

for element in my_list:
    print("Aktuelles Element aus der Liste 'my_list':", element)

````

Wir sehen, dass die Schleife automatisch beendet wird, sobald das letzte Element erreicht wurde.

> <span style="color:green"> **_AUFGABE:_** </span> Ersetze  `my_list` sowie den Iterator `element` mit eigenen Namen in der obenstehenden `for`-Schleife.

````python
names_list = ["Susanne", "Bernd", "Aurel", "Hamid"]

for participant in names_list:
    print("Am DSI Kurs nimmt teil:", participant)
````

Ein weiteres Beispiel:

````python
string1 = "Hallo"

for char in string1:
    print(char)
````

Neben Listen können wir ebenfalls über Dictionaries iterieren und die einzelnen Werte zu den jeweiligen Schlüsseln mit der `keys()`-Methode ausgeben lassen: 

> <span style="color:green"> **_AUFGABE:_** </span> Schreibe eine for-Schleife, die das Dictionary durchläuft und für jedes Land folgenden Satz auf der Konsole ausgibt:\
> „Die Hauptstadt von [Land] ist [Hauptstadt]“

````python
# Aufgabe: 
countries = {"Deutschland" : "Berlin",
             "Frankreich" : "Paris",
             "Polen" : "Warschau"}

for key in countries.keys():
    print(f"Die Hauptstadt von {key} ist {countries[key]}")

for value in countries.values():
    print(value)
````

Das gleiche Ergebnis kann man auch durch die `items()`-Methode erreichen:

````python
for key, value in countries.items():
    print(f"Die Hauptstadt von {key} ist {value}.")
````

Soll ein Codeabschnitt nur mehrfach ausgeführt werden, ohne dass dabei auf die Elemente einer Sequenz Bezug genommen wird, kann die `range()`-Funktion in Verbindung mit einer `for`-Schleife verwendet werden. Dies ist effizienter und besser lesbar, als beispielsweise extra eine Liste zu erstellen.

**Beispiel:**

````python
dummy_list = [0, 1, 2, 3, 4, 5, 6]

for i in dummy_list:
    print("Die Zeile wird 7 Mal ausgegeben.")

for i in range(7):
    print("Die Zeile wird 7 Mal ausgegeben.")





````

Wir sehen, dass wir die gleiche Ausgabe mit der `range()`-Funktion erhalten, ohne dass eine Sequenz wie `dummy_list` dafür erzeugt werden muss.

Da `i` als Zähler im Codeabschnitt nicht verwendet wird, ist das Erstellen einer Liste nicht notwendig, da ohnehin keine Elemente verarbeitet werden.

Darüber hinaus ist die Verwendung der `range()`-Funktion auch praktisch, wenn Zahlen iterativ verarbeitet werden sollen.

Als Beispiel sollen die ersten fünf Ganzzahlen aufaddiert werden und in `summe` gespeichert werden:

````python
summe = 0

for i in range(1, 6):
    summe = summe + i

print("Die Summe ist:", summe)



````

Der obige Ausdruck ist äquivalent zum nachfolgenden Code. Wir benutzen den sogenannten syntaktischen Zucker, um das Programm nach aktuellen Konventionen zu schreiben:

````python
summe = 0

for i in range(1, 6):
    summe += i

print("Die Summe ist:", summe)
````

Weiterhin kann die `range()`-Funktion dazu verwendet werden, um nicht nur die Elemente in einer Sequenz auszugeben, sondern auch ihren Index. Dies könnte man auf folgende Art lösen:

````python
my_list = ["A", "B", "C"]

for index in range(len(my_list)):
    print("Element:", my_list[index], "an Index:", index)
````

Für einen übersichtlicheren Code empfiehlt es sich, die `enumerate()`-Methode zu verwenden. Diese Funktion erwartet eine Liste als Argument und liefert für jeden Schleifendurchlauf das Element selbst *und* seinen Index zurückliefert. 

Für jede Iteration wird im nächsten Beispiel der Wert des Elements in `element` und sein Index in `index` gespeichert:

````python
for index, element in enumerate(my_list):
    print("Element an Index:", index, ":", element)
````

````python
countries = {"Deutschland" : "Berlin",
             "Frankreich" : "Paris",
             "Polen" : "Warschau"}

for index, key in enumerate(countries.keys()):
    print(f"Der Schlüssel {key} steht an Index {index}")
````

### 1.1) Zusammenfassung: Der richtige Umgang mit For-Schleifen in Python

````python
liste_namen = ["Anna"]

for name in liste_namen:
    print(f"Hallo {name}!")

for i in range(len(liste_namen)):
    print(f"Hallo {liste_namen[i]}!")

for index, name in enumerate(liste_namen):
    print(f"Hallo {liste_namen[index]}!")
````

### 1.2) `break`-Anweisung

In den Fällen, wo eine `for`-Schleife vorzeitig abgebrochen werden soll, kann die **`break`-Anweisung** verwendet werden.

Wird innerhalb einer Schleife ein `break` erreicht, wird die Schleife *sofort* verlassen.

Im folgenden Beispiel soll die `for`-Schleife verlassen werden, sobald das Schlüsselwort `stop` in der Liste erreicht wird:

````python
my_list = [1, 2, 3, "stop", 4, "stop"]

for element in my_list:
    if element == "stop":
        break        
    print(element)
````

Obwohl über die Elemente `element` in `my_list` iteriert wird, werden nur die ersten drei Elemente ausgegeben. Durch die `if`-Bedingung `element == "stop"` wird die Schleife verlassen, sobald das Element `stop` erreicht wird und die Iteration bei diesem Element abgebrochen.

> <span style="color:red"> **_HINWEIS:_** </span> Wenn das Schlüsselwort nicht ausgegeben werden soll, so ist die Reihenfolge der Codezeilen innerhalb der `for`-Schleife relevant.

````python
for element in my_list:
    if element == "stop":
        print(element)
        break
````

Wird der `print()`-Befehl vor der `break`-Anweisung ausgeführt, so wird das Schlüsselwort `"stop"` mit ausgegeben.

### 1.3) `continue`-Anweisung

Mit der `continue`-Anweisung können wir direkt zum Anfang der Schleife zurückkehren, ohne weiteren Code innerhalb der Schleife auszuführen.

Dies kann sich als nützlich erweisen, wenn wir nur an bestimmten Werten interessiert sind.

Im folgenden Beispiel werden nur ungerade Zahlen zwischen `1` und `10` ausgegeben, wird nämlich eine gerade Zahl erreicht, sorgt die `continue`-Anweisung dafür, dass der `print()`-Befehl nicht erreicht wird.

````python
for i in range(1, 11):  # Zahlen von 1 bis 10
    if i % 2 == 0:      # Erinnerung: Mit dem Modulo-Operator können wir überprüfen, ob eine Zahl gerade ist
        continue
    print(i, "ist eine ungerade Zahl.")
````

Im nächsten Beispiel werden alle Zahlen zwischen `1` und `20` ausgegeben, welche durch `3` teilbar sind:

````python
for i in range(1, 20):  # Zahlen von 1 bis 20
    if not i % 3 == 0:  # Erinnerung: Mit dem Modulo-Operator können wir überprüfen, ob eine Zahl ohne Rest teilbar ist
        continue
    else:
        print(i, "ist durch 3 teilbar.")
````

Wenn eine Zahl nicht durch 3 teilbar ist, wird durch die `continue`-Anweisung direkt wieder zu Beginn der Schleife gesprungen und der `print()`-Befehl wird in solch einem Schleifendurchlauf nicht ausgeführt.

### 1.4) `pass`-Anweisung

Mit der `pass`-Anweisung kann der Codeblock innerhalb einer Schleife leer bleiben, ohne dass der Ablauf des Programms gestört wird.

Dadurch kann die `pass`-Anweisung als vorübergehender Platzhalter eingesetzt werden, um einen leeren Codeblock zu kennzeichnen. Somit bleibt die Syntax gültig und der Programmfluss wird nicht durch eine Fehlermeldung unterbrochen, obwohl an dieser Stelle noch kein konkreter Code implementiert wurde.

Im folgenden Beispiel ist noch nicht klar, was passieren soll, wenn die`if`-Bedingung innerhalb der Schleife erfüllt ist.

````python
for i in range(25):
    if i % 5  == 0:
         pass # Wir wollen den Code für diese Bedingung irgendwann in der Zukunft hinzufügen

print("Das Ende des Programms wurde erreicht.")
````

Da Python-Code durch Einrückungen strukturiert ist, können wir nicht einfach keinen Code schreiben, sondern müssen mitteilen, dass dort in der Zukunft Code geplant ist. Dadurch erhalten wir eine Fehlermeldung, weil Python erwartet, dass auf eine `if`-Bedingung Code folgt.

Dieses Problem können wir mit der `pass`-Anweisung beheben:

````python
for i in range(25):
    if i % 5  == 0:
        pass  # Jetzt können wir das Programm ausführen, ohne dass ein Fehler erscheint!

print("Das Ende des Programms wurde erreicht.")
````

## 2) `while`-Schleifen

Im Gegensatz zu `for`-Schleifen wird eine `while`-Schleife beendet, wenn eine Bedingung *nicht mehr* erfüllt ist, bzw. Code ausführt, *solange* eine Bedingung erfüllt ist. 

```python
while Bedingung:
    Code, der ausgeführt werden soll, solange Bedingung erfüllt ist
```

**Beispiele:**

````python
counter = 0

while counter < 5:
    print("Der aktuelle Wert von counter ist", counter)
    counter = counter + 1



````

Sobald `counter` den Wert `8` erreicht, ist die Bedingung nicht mehr erfüllt, was zum Verlassen der Schleife nach dem aktuellen Durchlauf führt.

> <span style="color:red"> **_HINWEIS:_** </span> Auch innerhalb von `while`-Schleifen spielt die Reihenfolge der Codezeilen eine Rolle.

````python
counter = 0

while counter < 6:
    print("Der aktuelle Wert von counter ist", counter)
    counter = counter + 1
````

Wird der `print()`-Befehl vor der Inkrementierung von `counter` ausgeführt, so wird `8` ebenfalls mit ausgegeben. Von der Lesbarkeit her ist es empfehlenswert, wie in der ersten Variante zunächst den `print()`-Befehl auszuführen, weil durch die Bedingung `counter < 8` signalisiert wird, dass nur die Zahlen kleiner als `8` ausgegeben werden sollen.

> <span style="color:red"> **_HINWEIS:_** </span> Variablen wie `counter` werden häufig als Hilfsmittel zur Steuerung von `while`-Schleifen verwendet. Wichtig dabei ist, dass sie *vor* der `while`-Schleife initialisiert werden.

Im folgenden Beispiel wird ein Nutzer mithilfe der `input()`-Funktion so lange um eine Eingabe gebeten, bis diese korrekt ist.

````python
user_input = ""

while user_input != "A" and user_input != "a":
    user_input = input("Was ist der erste Buchstabe des Alphabets?\n")

print("Korrekt!")
````

Wir sehen hier auch eine etwas komplexere Bedingung in der `while`-Schleife.  Der Teil in der Klammer ist `True`, wenn die Nutzereingabe `A` oder `a` ist.  Dies wird durch `not` verneint, d.h. die Schleife wird solange durchlaufen, wie die Eingabe *nicht* `A` oder `a` lautet.

### 2.1) Endlosschleifen

> <span style="color:red"> **_HINWEIS:_** </span> Wir müssen aufpassen, dass wir keine **Endlosschleife** erzeugen, in der die Bedingung immer erfüllt ist. Diese Schleifen werden *niemals* verlassen werden und laufen daher unendlich lange.   

**Beispiel:**

Im folgenden Programm wird die Schleife verlassen, wenn `x` kleiner als `5` ist. Dies ist jedoch *nie* der Fall, da `x = 2` innerhalb der Schleife nicht verändert wird. Dadurch enthält der Beispielcode eine Endlosschleife. Eine Ausführung kann dazu führen, dass die Entwicklungsumgebung hängen bleibt und neu gestartet werden muss. Führe daher folgenden Code am besten **nicht** aus und lasse ihn kommentiert!

Die `break`-Anweisung ermöglicht zusammen mit einem Zähler `i` eine Art Stoppmechanismus, um eine `while`-Schleife nach einer bestimmten Anzahl von Iterationen zu beenden, wodurch ungewollte Endlosschleifen vermieden werden können.

**Beispiel:**

````python
i = 0  # Eine Hilfsvariable i wird definiert, um die Iterationen zu zählen
x = 2

while i < 10:
    print(x)
    i = i + 1

    if i == 10:
        break

````

Im obigen Beispiel wird die `while`-Schleife nach `i = 10` Iterationen beendet und führt somit zur keinen Endlosschleife.

Das nächste Beispiel zeigt eine ebenfalls eine potentielle Endlosschleife, deren Abbruchbedingung niemals `False` wird. Dennoch können wir die Schleife verlassen durch eine `break`-Anweisung:

````python
while True:
    user_input = input("Schleife beenden? j/n")

    if user_input == "j":
        print("Die Schleife wird nun beendet.")
        break

    print("Die Schleife läuft noch!")

print("Schleife wurde verlassen.")
````
