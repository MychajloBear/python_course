Syntaktischer Zucker ist ein Begriff aus der Informatik und bezieht sich auf eine bestimmte Art von Sprachkonstruktionen in Programmiersprachen. Es handelt sich dabei um eine Syntax, die speziell dazu entwickelt wurde, häufig verwendete und repetitive Konstruktionen (wie die obige) in einer Programmiersprache auf eine kürzere, intuitivere und leichter les- und schreibbare Weise auszudrücken.

Die obigen Symbole fügen Python keine neue Funktionalität hinzu, sondern nur eine neue Syntax. Schauen wir uns den obigen Beispielcode mit der `+=` Schreibweise an:

Mit **Slicing** können wir gleichzeitig auf mehrere Elemente eines Strings `my_string` zugreifen. Dabei entstehen sogenannte **Teilstrings** (englisch: *substrings*):

- `my_string[start:stop]`: Gibt die Elemente von Index `start` bis `stop - 1` zurück.
- `my_string[start:]`: Gibt die Elemente von Index `start` bis zum Ende des Strings zurück.
- `my_string[:stop]`: Gibt die Elemente vom Anfang des Strings bis zum Index `stop - 1` zurück.

> <span style="color:red"> **_HINWEIS:_** </span> Beim Slicing werden bei Angabe von `stop` nur die Elemente bis zur Position `stop - 1` berücksichtigt.

## 6) Numerische Datentypen: Integers & Floats
Python unterscheidet im Wesentlichen zwei Arten von Zahlen:

- **Integers (int)**: Ganzzahlen ohne Dezimalstellen.

- **Floats (float)**: Fließkommazahlen mit Nachkommastellen.

> <span style="color:red"> **_HINWEIS:_** </span> Als Dezimaltrennzeichen wird im Code immer der englische Punkt (.) verwendet.

**Beispiele:**

```python
print(2+3)     # Datentyp: Integer (2,3, sowie 5)
print(4-1)
print(2*3)
print(4/2)     # Datentyp: Integer (2,4), aber Float (2.0)

print(2^3)     # Potenzieren (2 hoch 3 = 8).
```

### 6.1) Die Modulo-Funktion `%`

Der Modulo-Operator ist ein klassisches und extrem wichtiges Werkzeug in der Programmierung. Er teilt zwei Zahlen und gibt ausschließlich den Rest dieser Division zurück:

```python
print(4%2)       # 
print(4%3)       # 
print(4%4)       # 
print(4%5)       # 
```

### 6.2) Typische Anwendungsfälle für Modulo

**A) Prüfung auf Teilbarkeit (Gerade / Ungerade)**
Eine Zahl ist genau dann gerade, wenn die Division durch 2 keinen Rest hinterlässt (also ``Rest == 0``).

```python
print(16 % 2)  # Ausgabe: ? -> Zahl ist ?
print(17 % 2)  # Ausgabe: ? -> Zahl ist ?
```

**B) Zyklische Abläufe (Wiederkehrende Muster)**

Mit Modulo lassen sich Prozesse abbilden, die nach dem Erreichen eines Limits wieder bei 0 beginnen müssen (wie z. B. die Stunden einer Uhr).

**Beispiel:**

```python
hour = 22                                      # Uhrzeit 22:00 
hours_to_add = 5                               # 5 Stunden werden addiert
new_hour = (hour + hours_to_add) % 24          # Modulo 24 für 24 Stunden am Tag

print(new_hour)                                # 3:00, da 5 Stunden vergangen sind
```

## 6) Einwertige Datentypen - Zusammenfassung

*Jede* Variable in Python hat, neben einem Wert auch einen **Datentyp**. Der Datentyp bestimmt, welche *Art* von Daten in der Variable gespeichert werden, z.B. Zahlen, Texte, usw. Bei **einwertigen** (oder auch **primitiven**, bzw. **elementaren**) Datentypen wird jeder Variable genau *ein* einzelner Wert zugewiesen. Die meisten haben wir bereits kennengelernt, z.B. Strings, Booleans, Integers und Floats.

| **Datentyp** | **Bezeichnung** | **Beschreibung**   | **Beispiele**            |
|--------------|-----------------|--------------------|--------------------------|
| String     | `str`             | Text               | `"Hallo Welt!"`          |
| Boolean    | `bool`            | Wahrheitswert      | `True`, `False`          |
| Integer    | `int`             | Ganzzahl           | `13, 28, -7`             |
| Float      | `float`           | Fließkommazahl     | `4.3, -3.14, 1045.2`     |
| None       | `None`            | "Nichts"           | `None`                   |
