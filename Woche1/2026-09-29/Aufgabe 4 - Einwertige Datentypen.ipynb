### Aufgabe 1: Datentypüberprüfung

Überlege dir das Ergebnis der folgenden `print()`-Befehle *bevor* du die folgenden Zellen ausführst und schreibe deine Vermutung mit einer Begründung als Kommentar ans Ende jeder Zeile.

```python
zahl = "vier"
number_string = 5

print(type(zahl))           # <class 'str'>  (Text)
print(type(number_string))  # <class 'int'>  (Integer)
```

```python
x = 5
y = 10
z = 7.2

print(type(x + y))  # <class 'int'>   (5 + 10 = 15; Integer)
print(type(x - z))  # <class 'float'> (5 - 7.2 = -2.2; Float-Zahl)
print(type(x / y))  # <class 'float'> (5 / 10 = 0.5; Ergebnis von / ist immer eine Float-Zahl)
```

```python
a = True   # wie 1
b = False  # wie 0
c = 3

print(type(a + b))  # <class 'int'>   (1 + 0 = 1; Integer)
print(type(a - b))  # <class 'int'>   (1 - 0 = 1; Integer)
print(type(a / c))  # <class 'float'> (1 / 3 = 0.333...; Ergebnis von / ist immer eine Float-Zahl)
```

### Aufgabe 2: Interaktiver Taschenrechner

Schreibe ein Programm, welches:

- zwei ganze Zahlen von einem Benutzer anfordert und diese in den Variablen `a` und `b` speichert,
- die Variablen `a` und `b` in Integers umwandelt,
- direkt die Summe dieser Zahlen berechnet, ohne eine zusätzliche Variable zu definieren,
- `f`-Strings verwendet, um das Ergebnis auszugeben.

**Tipp:** Die `int()`-Funktion könnte dabei hilfreich sein.

```python
a = int(input("Zahl 1:"))
b = int(input("Zahl 2:"))
print(f"{a} + {b} = {a + b}") 
```

### Aufgabe 3: Arithmetisches Mittel berechnen

Schreibe ein Programm, das zwei ganze Zahlen `a` und `b` als Eingabe anfordert und das arithmetische Mittel `(a + b) / 2` dieser Zahlen ausgibt.

**Tipp:** Wandle die Eingaben vor der Berechnung in Integers um.

```python
a = int(input("Geben Sie die erste Zahl (a) ein: "))
b = int(input("Geben Sie die zweite Zahl (b) ein: "))

result = (a + b) / 2
print(result)
```

### Aufgabe 4: Tage in Sekunden umrechnen

Schreibe ein Programm, das eine Anzahl von Tagen als Eingabe anfordert und die entsprechende Dauer in Sekunden ausgibt.

```python
days = int(input("Geben Sie die Anzahl der Tage ein: "))

seconds = days * 24 * 60 * 60
print(seconds)
```

### Aufgabe 5: Dollar in Euro umrechnen

Schreibe ein Programm, das nach einem Betrag in Dollar fragt, diesen Betrag in Euro umrechnet und auf zwei Nachkommastellen gerundet ausgibt.

Der angenommene Umrechnungskurs lautet: `$1.00 = 0.92€`.

**Tipp:** Weil eine beliebiger Betrag angegeben werden kann, sollte die Eingabe in eine Fließkommzahl umgewandelt werden.

```python
dollar = float(input("Geben Sie den Betrag in Dollar ein: "))

euro = round(dollar * 0.92, 2)
print(euro)
```

### Aufgabe 6: Berechnung der Kreisfläche

Schreibe ein Programm, welches den Radius `r` eines Kreises als Eingabe anfordert, die Fläche des Kreises näherungsweise mit der Formel: 

`area = 3.14 * r ** 2` 

berechnet und das Ergebnis auf `4` Nachkommastellen gerundet ausgibt.

```python
r = float(input("Geben Sie den Radius (r) ein: "))

area = round(3.14 * r ** 2, 4)
print(area)
```
