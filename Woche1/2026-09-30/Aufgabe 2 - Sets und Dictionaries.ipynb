## Mengen

### Aufgabe 1: Menge erstellen, Elemente hinzufügen & entfernen

- Erstelle eine Menge `colors` mit drei unterschiedlichen Farben (z.B. `blue`).
- Füge nach der Initialisierung noch eine weitere Farbe dazu.
- Entferne danach eine der 4 Farben in `colors`.
- Lasse dir nach jedem Schritt die Menge ausgeben.

**Tipp:** Achte darauf, welche Art von Klammern bei Mengen verwendet werden.

```python
colors = {"red", "green", "blue"}
print(colors)

colors.add("yellow")
print(colors)

colors.remove("red")
print(colors)
```

### Aufgabe 2: Menge ergänzen

Ergänze die obige Menge mit der Menge `colors2`. Gib anschließend das durch die neuen Farben ergänzte Set aus.

**Tipp:** Die beide Mengen sollen vereinigt werden.

```python
# Lasse den folgenden Code unverändert und führe nur die Zelle aus

colors2 = {'pink', 'purple', 'orange'}
```

```python
colors.update(colors2)
print(colors)
```

### Aufgabe 3: Gemeinsame Elemente

Speichere die gemeinsame Elemente aus dem Schnitt der obigen Menge mit `new_colors` in `common_elements`.

```python
# Lasse den folgenden Code unverändert und führe nur die Zelle aus

new_colors = {'red', 'blue', 'orange'}
```

```python
common_elements = colors.intersection(new_colors)
print(common_elements)
```

### Aufgabe 4: Einzigartige Elemente

- Finde die Elemente, die nur in der Menge `colors` und nicht in `new_colors` sind.
- Finde die Elemente, die nur in der Menge `new_colors` und nicht in `colors` sind.

**Tipp:** Die Reihenfolge ist hierbei entscheidend.

```python
only_in_colors = colors.difference(new_colors)
only_in_new_colors = new_colors.difference(colors)
print(only_in_colors)
print(only_in_new_colors)
```

## Dictionaries

### Aufgabe 1: Dictionary erstellen

- Erstelle ein Wörterbuch `dog` für einen Hund mit den Schlüsseln: `name` (String), `breed` (String), `age` (Integer). Achte auf die Verwendung der richtigen Datentypen und wähle plausible Werte.
- Gib das Dictionary aus.
- Verifiziere anschließend den Datentyp des Dictionaries.

```python
dog = {
    "name": "Rex",
    "breed": "German Shepherd",
    "age": 3
}

print(dog)
print(type(dog))
```

### Aufgabe 2: Neue Key-Value-Paare hinzufügen & ändern

- Füge nach der Initialisierung den Schlüssel `good_dog` (Boolean) mit dem Wert `True` ein.
- Überschreib den Wert für `age` mit `5`. 
- Gib anschließend das Dictionary erneut aus.

```python
dog["good_dog"] = True
dog["age"] = 5

print(dog)
```

### Aufgabe 3: Schlüsselinformationen ausgeben

- Ermittle mithilfe des `len()`-Befehls, wie viele Key-Value Paare es in `dog` gibt.
- Lass dir anschließend alle Schlüssel ausgeben.
- Lasse dir abschließend alle Werte ausgeben.

```python
print(len(dog))
print(dog.keys())
print(dog.values())
```
