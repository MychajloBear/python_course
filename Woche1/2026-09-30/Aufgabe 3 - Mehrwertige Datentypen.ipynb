### Aufgabe 4: Gemeinsamkeiten und Unterschiede von Schülerkursen

- Gib alle Schüler aus, die sowohl Mathe als auch Informatik belegen.
- Gib alle Schüler aus, die nur Mathe belegen.
- Gib alle Schüler aus, die nur Informatik belegen.
- Gib alle Schüler aus, die mindestens einen der beiden Kurse belegen.
- Gib die Anzahl der Schüler aus, die mindestens einen Kurs belegen.


```python
# Lasse den folgenden Code unverändert und führe nur die Zelle aus

mathe = {"Anna", "Ben", "Clara", "David", "Eva"}
informatik = {"Ben", "David", "Felix", "Greta", "Anna"}

# Gib alle Schüler aus, die sowohl Mathe als auch Informatik belegen
print(mathe & informatik)

# Gib alle Schüler aus, die nur Mathe belegen
print(mathe - informatik)

# Gib alle Schüler aus, die nur Informatik belegen
print(informatik - mathe)

# Gib alle Schüler aus, die mindestens einen der beiden Kurse belegen
print(mathe | informatik)

# Gib die Anzahl der Schüler aus, die mindestens einen Kurs belegen
print(len(mathe | informatik))
```
