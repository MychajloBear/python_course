## 1) Funktionen

Bisher liefen unsere Programme immer von oben nach unten ab. Oft benötigen wir jedoch dieselbe Funktionalität an *mehreren* Stellen im Programm. Beispielsweise könnte es nötig sein, eine Person mehrfach zu begrüßen oder an unterschiedlichen Punkten den Mittelwert zweier Zahlen zu berechnen. Es wäre sehr unpraktisch und ineffizient, denselben Code dafür jedes Mal erneut zu schreiben, da dies das Programm unnötig aufblähen würde. Eine elegantere Lösung bieten **Funktionen**. Funktionen erfüllen spezifische Aufgaben und können von *überall* im Programm aufgerufen werden. Sie ermöglichen es, denselben Code mehrfach zu verwenden, ohne ihn jedes Mal neu schreiben zu müssen.

Wir haben Funktionen bereits an mehreren Stellen kennengelernt, zum Beispiel die eingebaute `print()`-Funktion in Python. Nun wollen wir lernen, wie man *eigene* Funktionen definiert und nutzt.

Generell ist die Syntax von Funktionen folgendermaßen aufgebaut:

```python
def funktion_name(parameter1, parameter2, ...):
    anweisung1
    anweisung2
    anweisung3
```

- `funktion_name` steht für den **Namen** der Funktion, über den sie aufgerufen werden kann. Ähnlich wie bei Variablen sollten Namen gewählt werden, welche die Aufgabe der Funktion klar beschreiben.
- Hinter dem Funktionsnamen stehen in runden Klammern beliebig viele **Parameter**, über die der Funktion beim Aufruf konkrete Werte übergeben werden können. Parameter sind Variablen in der Funktionsdefinition, die als Platzhalter für die Werte dienen, welche die Funktion beim Aufruf erhält. Sie legen fest, welche **Eingabewerte** die Funktion erwartet. Hingegen sind **Argumente** die tatsächlichen Werte, die beim Aufruf einer Funktion übergeben werden. Diese Werte ersetzen die Parameter innerhalb der Funktion, sodass mit ihnen gearbeitet werden kann.
- Der Codeblock der Funktion folgt in den nächsten Zeilen nach dem Doppelpunkt und muss genauso wie bei Verzweigungen und Schleifen *eingerückt* sein.

## 2) Type Annotations

**Type Annotations** können verwendet werden, um die Lesbarkeit, Wartbarkeit und manchmal auch die Performance deiner Programme zu verbessern.

Mit Type Annotations kannst du den Datenyp von Variablen, Funktionsargumenten und Rückgabewerten spezifizieren. Das hilft anderen Entwicklern (und dir selbst nach einer langen Pause) zu verstehen, was dein Code macht, ohne ihn ausführen zu müssen. Die Verwendung ist dabei optional. Ähnlich wie bei Kommentaren werden Type Annotations von Python nicht weiter beachtet. 

Während (mehrzeilige) Kommentare hervorragend für ausführlichere Erklärungen und Dokumentation geeignet sind, bieten Type Annotations eine *schnelle* und *unmittelbare* Art, Informationen bezüglich Datentypen bereitzustellen.

Das Schreiben von Type Annotations ist dabei oft schneller als das Verfassen umfangreicher Dokumentatonen, insbesondere bei der Entwicklung neuer Funktionen oder beim Prototyping. Sie ermöglichen es dir, die Typsicherheit deines Codes schnell zu verbessern, ohne den Fluss deiner Gedanken durch das Verfassen detaillierter Erklärungen unterbrechen zu müssen.

Type Annotations werden direkt im Code integriert, indem nach Variablen oder Funktionsparametern ein Doppelpunkt `:` und der erwarteten Datentyp angegeben wird. 

Bei zusammengesetzten Datentypen können die Annotationen allgemein gehalten werden, in dem z.B. nur `list` für eine erwartete Liste angegeben wird. Falls bereits bekannt, kann auch der genaue Datentyp der einzelnen Elemente angegeben werden, wie beispielsweise `list[float]` oder `dict[str, int]`. 

Im folgenden Beispiel werden einwertige Variablen mit den Datentypen `int` für ganze Zahlen, `str` für Strings und `bool` für Wahrheitswerte annotiert:

````python
# Hier kommt dein Code hin


print(number)
print(name)
print(is_prime)
````

Wir sehen, dass die Annotation keinen Einfluss auf den Wert einer Variable hat. 

> <span style="color:red"> **_HINWEIS:_** </span> Es sollte beachtet werden, nicht aus Versehen den falschen Datentyp zu annotieren, was zwar keine Fehlermeldung verursacht, jedoch für Verwirrung sorgen kann.

**Beispiel:**

````python
# Hier kommt dein Code hin
number: int = 5.0
name: str = False
is_prime: bool = 1234

print(number)
print(name)
print(is_prime)
````

Genauso können auch mehrwertige Datentypen wie Listen, Dictionaries und Tuples annotiert werden:

````python
friends: list[str] = ["Lena", "Max", "Sophie"]
age_of_friends: dict[str, int] = {"Lena": 28, 
                                  "Max": 32, 
                                  "Sophie": 24}
best_friends: set[str] = {"Lena", "Sophie"}
````

Bei Funktionen werden die Datentypen der Parameter mit einem Doppelpunkt `:` und der Datentyp der Rückgabewerte mit einem ein Pfeil `->` annotiert.

**Beispiele:**

````python
def add_numbers(a: int, b: int) -> int:  # Argumente: Integers, Rückgabewert: Integer
    return a + b

def create_full_name(first_name: str, last_name: str) -> str:  # Argumente: Strings, Rückgabewert, String
    return first_name + " " + last_name

def is_even(number: int) -> bool:  # Argument: Integer, Rückgabewert: Boolean
    return number % 2 == 0

def get_max_value(numbers: list[int]) -> int:  # Argument: Liste von Integers, Rückgabewert: Integer
    return max(numbers) 

def show_all_friends(friends: list[str]) -> None: # Argument: Liste von Strings, Rückgabewert: None
    for friend in friends:
        print(friend)

def add_dog(dogs: dict[str, str], name: str, breed: str) -> dict[str, str]:  # Argumente: Dictionary von Strings, String, String
    dogs[name] = breed                                                       # Rückgabewert: Dictionary von Strings
    return dogs
````

Das bietet sich insbesondere dann an, wenn der Code innerhalb der Funktion sonst nicht korrekt funktionieren würde, etwa wenn datentypspezifische Methoden verwendet werden.

# 3) Docstrings

Bei größeren Pythonprojekten ist eine Dokumentation des Codes sehr wichtig. Besonders bei komplexen Funktionen und längerem Code kann man leicht den Durchblick verlieren. Eine Dokumentation hilft auch anderen Programmierern dabei, deinen Code zu verstehen.

**Docstrings** bieten eine einfache Möglichkeit einen Teil des Programmes, sei es eine Funktion, eine Klasse oder ein Modul, zu dokumentieren.

Ein Docstring wird durch drei Einzel- `'''` oder Doppelanführungszeichen `"""` definiert.

Wenn man eine Funktion dokumentiert, gibt man als erstes eine kleine Beschreibung der Funktion an. Als nächstes spezifiziert man die Argumente welche die Funktion erwartet und das Ergebnis, welches von der Funktion zurückgeliefert wird. Dadurch wird anderen Programmierern die Funktionsweise der Funktion schneller klar, als wenn sie *nur* den Code zur Verfügung hätten. Unten siehst du ein kurzes Beispiel, wie für eine (einfache) Funktion der Docstring aussehen kann:

````python
def add_numbers(x, y):
    """
    Liefert die Summe der zwei Zahlen.

        Parameters:
              x (int): Erste Zahl
              y (int): Zweite Zahl

        Returns:
            sum (int): Die Summe der beiden Zahlen x und y.
    """
    return x + y
````

Diesen Docstring kann man im Code der Funktion sehen, wenn es unklar ist was für Input oder Ouput ein Funktion hat oder um schnell zu verstehen was die Funktion macht.

Zusätzlich findet sich der Docstring auch unter dem Attribut `__doc__` der Funktion als String.

````python
print(add_numbers.__doc__)
````

Das gleiche Ergebnis erhalten wir auch über die Built-In Funktion `help`:

````python
help(add_numbers)
````

Docstrings kannst du in allen großen Projekten (manchmal mit unterschiedlichen Dokumentationsstil) beobachten. Unten ist z.B. eine Funktion aus dem [Seaborn Code](https://github.com/mwaskom/seaborn/blob/master/seaborn/utils.py) gezeigt.

## 4) Prinzipien

Neben der reinen Syntax gibt es einige wichtige Aspekte, die man beim Schreiben von Funktionen berücksichtigen sollte. Daher wollen wir nun einige Prinzipien für "gute" Funktionen erkunden:

> **Prinzip Nr. 1: Eine Funktion sollte nur eine einzige Aufgabe erfüllen.**

Das folgende Beispiel zeigt eine Funktion `get_smallest_element_and_average()`, die *zwei* Aufgaben übernimmt: Sie ermittelt sowohl den Durchschnittswert, als auch das kleinste Element in einer Liste. Beide Berechnungen sind in einer einzigen Funktion zusammengefasst, was nicht ideal ist. Falls wir in einem anderen Kontext nur die Durchschnittsberechnung benötigen, dann leistet die Funktion unnötigerweise mehr als erforderlich:

````python
def get_smallest_element_and_average(lst: list[int]) -> tuple[int, float]:
    smallest = min(lst)
    average = sum(lst) / len(lst)
    return smallest, average
````

````python
my_list = [1, 2, 3, 1, 0, 2]

smallest_element, average = get_smallest_element_and_average(my_list)

print("Das kleinste Element der Liste ist:", smallest_element)
print("Der Mittelwert der Liste ist:", average)
````

Aufgrund der Initialisierung von `smallest_element` und `average` durch die Rückgabewerte innerhalb einer Zeile wird zudem der Code unübersichtlicher. Besser ist es daher, für jeden Aufgabenteil *eine* Funktion zu schreiben. Diese Funktionen sind aufgrund der klaren Aufgabe leichter zu lesen und einzusetzen.

> **Prinzip Nr. 3: Eine Funktion sollte klar nach bestehenden Konventionen benannt werden.**

Ein gut gewählter Funktionsname kann die Lesbarkeit und Wartbarkeit von Code erheblich verbessern. Dabei hilft oft ein Verb zu Beginn der Funktionsbezeichnung. Im Folgenden sind einige bewährte Praktiken für die Benennung von Funktionen zusammen mit Beispielen aufgeführt:

- **`get_`** für Funktionen, die Werte zurückgeben, ohne diese zu verändern, z.B. `def get_average(numbers):`

- **`is_`** oder **`has_`** für Funktionen, die einen booleschen Wert zurückgeben, um eine Bedingung zu prüfen, z.B. `def is_prime(number):`
  
- **`calculate_`** für Funktionen, die eine Berechnung durchführen und das Ergebnis zurückgeben, z.B. `def calculate_interest(rate, principal, time):`

## 5) Namensräume

Um zu verstehen, wie Python den Zugriff auf Variablen handhabt und was passiert, wenn in verschiedenen Funktionen gleichnamige Variablen verwendet werden, ist es hilfreich, das Konzept von **Namensräumen** (englisch: *namespaces*) zu betrachten. Ein Namensraum ist ein Bereich innerhalb eines Programms, in dem ein Name (z.B. einer Funktion oder einer Variablen) gültig und ansprechbar ist.

Hierfür schauen wir uns den **eingebauten** Namensraum (englisch: *built-in namespace*), den **globalen** Namensraum (englisch: *global namespace*) und den **lokalen** Namensraum (englisch: *local namespace*) genauer an.

In diesem Programm wurde die Variable `x` zweimal definiert: Einmal innerhalb der Funktion `print_local_x()` und einmal außerhalb. Wie entscheidet Python, auf welches `x` zugegriffen wird?

Der Zugriff auf Variablen und die Sichtbarkeit von Funktionen in Python werden durch Namensräume bestimmt. Namensräume helfen dabei, Konflikte zu vermeiden, die auftreten können, wenn verschiedene Module dieselben Namen für Variablen verwenden. 

Python unterscheidet zwischen drei Arten von Namensräumen:
    
- Der **eingebaute Namensraum** enthält alle eingebauten Objekte in Python und ist im gesamten Programm verfügbar. Dazu gehören Funktionen wie `print()` oder `abs()`, die bereits definiert sind, bevor eigener Code ausgeführt wird.
- Der **globale Namensraum** bezieht sich auf die Objekte bzw. Variablen, die außerhalb von Funktionen definiert sind und ist innerhalb eines Programms überall zugänglich.
- Der **lokale Namensraum** umfasst Namen, die *innerhalb* einer Funktion definiert sind und nur in diesem lokalen Kontext verfügbar sind. *Jede* Funktion hat also ihren eigenen Namensraum.

Der eingebaute Namensraum umfasst den globalen Namensraum, der wiederum den lokalen Namensraum einschließt. Dabei ist der Zugriff von äußeren auf innere Namensräume nicht möglich; beispielsweise können Variablen aus lokalen Namensräumen nicht im globalen Namensraum angesprochen werden.

### 1.3) Lokaler Namensraum

Der lokale Namensraum wird erzeugt, wenn eine Funktion aufgerufen wird und existiert bis zu dem Zeitpunkt, an dem die Funktion wieder verlassen wird.

````python
# Hauptprogramm: Alles hier definierte ist im globalen Namensraum
var = "Globale Variable"

def namespace_demo():
    # Lokaler Namensraum der Funktion namespace_demo() beginnt hier
    var = "Lokale Variable"  # var wird lokal definiert und ändert die globale Variable var NICHT.

    print("Der Wert der lokalen Variable var ist:", var)
    # Lokaler Namensraum der Funktion namespace_demo endet hier.

namespace_demo()

print("Der Wert der globalen Variable var ist weiterhin:", var)
````

Aus diesem Beispiel wird deutlich, dass die jeweilige lokale und globale Variable `var` unabhängig voneinander sind und ihre jeweiligen Werte beibehalten. Trotz des Zugriffs auf `var` innerhalb von `namespace_demo()` und des scheinbaren Versuchs, ihren Wert zu ändern, bleibt die globale Variable unverändert.
