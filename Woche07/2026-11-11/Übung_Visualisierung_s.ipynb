```
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Fahrradverleih.csv")
print(df.head())
print(df.info())
```

## Aufgabe 2 Antworten
Typischerweise ca. 8-10 Fahrten pro Monat. Häufige Werte liegen im Bereich 7-10. Hohe Werte (15-17) können als potenzielle Ausreißer betrachtet werden.

```
sns.histplot(data=df,x="Anzahl_Fahrten_Monat",bins=17)
plt.title("Verteilung der monatlichen Fahrten")
plt.ylabel("Häufigkeit")
plt.show()
```

```
sns.boxplot(data=df,x="Fahrtdauer")
plt.xlabel("Fahrtdauer (Minuten)")
plt.show()
```

```
sns.countplot(data=df,x="Startstation")
plt.title("Nutzung der Startstationen")
plt.show()
```

```
sns.barplot(data=df,x="Kundentyp",y="Preis")
plt.title("Durchschnittlicher Preis nach Kundentyp")
plt.show()
```

```
sns.countplot(data=df,x="Kundentyp",hue="Bewertung")
plt.show()
```

```
corr=df.select_dtypes(include="number").corr()
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.show()
```

```
sns.scatterplot(data=df,x="Distanz",y="Preis",hue="Kundentyp")
plt.show()
```

```
g=sns.FacetGrid(df,col="Kundentyp")
g.map(sns.histplot,"Preis")
plt.show()
```

```
order=["Mo","Di","Mi","Do","Fr","Sa","So"]
sns.boxplot(data=df,x="Wochentag",y="Fahrtdauer",order=order)
plt.show()
```
