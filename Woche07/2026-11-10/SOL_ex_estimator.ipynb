## Konfidenzintervall
#### Beispiel:
Szenario: Auf einer Milchpackung steht ein Nennwert von 1000 ml. Wir wissen aus Erfahrung, dass die Abfüllmaschine eine bekannte Standardabweichung von $\sigma = 5$ ml hat. Wir wollen nun schätzen, wo der wahre Mittelwert der Produktion liegt.

* Daten der Stichprobe ($n=10$): Die Messwerte (in ml) sind: $998, 1002, 995, 1005, 999, 1001, 1000, 992, 1008, 997$
* 1. Berechnung der Kennzahlen:
    * Stichprobenmittelwert ($\bar{x}$):$$\bar{x} = 999,7 \text{ ml}$$

    * Standardfehler ($SE$): Da $\sigma$ bekannt ist, berechnet sich der Standardfehler der Mittelwerte als:
$$SE = \frac{\sigma}{\sqrt{n}} = \frac{5}{\sqrt{10}} \approx 1,58$$
* Bestimmung des Konfidenzintervalls (95 %): Wir wählen ein Vertrauensniveau von 95 %. Da es sich um ein zweiseitiges Intervall handelt, nutzen wir den Z-Wert für
$$1 - \alpha/2 = 0,975 \rightarrow z = 1,96$$
* Die Formel:
$$
KI = \bar{x} \pm z \cdot \frac{\sigma}{\sqrt{n}}
$$

$$
KI = 999,7 \pm 1,96 \cdot 1,58
$$

$$
KI = 999,7 \pm 3,1
$$
* Ergebnis:$$[996,6 \text{ ml}; 1002,8 \text{ ml}]$$
* Interpretation: Wir sind uns zu 95 % sicher, dass der wahre Durchschnittswert $\mu$ aller produzierten Milchkartons in diesem Bereich liegt. Da die 1000 ml im Intervall enthalten sind, scheint die Maschine korrekt eingestellt zu sein.

```python
import numpy as np
from scipy.stats import norm

# 1. Daten und Parameter
daten = [998, 1002, 995, 1005, 999, 1001, 1000, 992, 1008, 997]
sigma = 5  # Bekannte Standardabweichung der Grundgesamtheit
n = len(daten)
vertrauen = 0.95

# 2. Berechnungen
mittelwert = np.mean(daten)
standard_fehler = sigma / np.sqrt(n)

# Z-Wert für 95% (zweiseitig)
# norm.ppf berechnet das Quantil (0.975 für 95% Vertrauen)
z_wert = norm.ppf(1 - (1 - vertrauen) / 2)

# Fehlerspanne (Margin of Error)
fehler = z_wert * standard_fehler

# Intervall Grenzen
untergrenze = mittelwert - fehler
obergrenze = mittelwert + fehler

# 3. Ausgabe
print(f"Stichprobenmittelwert: {mittelwert:.2f} ml")
print(f"Standardfehler (SE): {standard_fehler:.2f}")
print(f"Z-Faktor: {z_wert:.3f}")
print(f"95% Konfidenzintervall: [{untergrenze:.2f} ml, {obergrenze:.2f} ml]")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))
tüten_index = np.arange(1, n + 1)

# Alle Messpunkte einzeichnen
plt.scatter(tüten_index, daten, color='royalblue', label='Einzelne Messwerte', zorder=5)
# Whiskers (Fehlerbalken) um jeden Punkt zeichnen
plt.errorbar(tüten_index, daten, yerr=fehler, fmt='none', ecolor='lightskyblue',
             capsize=5, alpha=0.6, label='Unsicherheits-Marge (95%)')
plt.axhline(mittelwert, color='red', lw=2, label=f'Stichproben-Mittelwert: {mittelwert:.1f} ml',
            linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

#### 🔋 Beispiel: Akku-Stresstest für das „Model X“

Wir testen eine Stichprobe von **$n = 64$** Smartphones eines neuen Typs unter Volllast, um die durchschnittliche Akkulaufzeit der gesamten Serie zu schätzen.

##### 1. Die Ausgangslage (Stichproben-Ergebnisse)
* **Durchschnittliche Laufzeit ($\bar{x}$):** $12$ Stunden
* **Stichprobenstandardabweichung ($s$):** $2,4$ Stunden

##### 2. Der ZGS-Joker (Zentraler Grenzwertsatz)
Da unsere Stichprobe mit **$n = 64$** deutlich über der Grenze von $30$ liegt, greift der **Zentrale Grenzwertsatz**. Selbst wenn die Laufzeiten einzelner Akkus völlig wild (nicht normalverteilt) streuen, verhält sich der *Mittelwert* der Stichprobe wie ein braver Soldat in einer Normalverteilung.

##### 3. Die Berechnung des 95 % Konfidenzintervalls
Um den Bereich für den wahren Mittelwert der Grundgesamtheit $\mu$ zu finden, berechnen wir:

* **Standardfehler ($SE$):**
    $$SE = \frac{s}{\sqrt{n}} = \frac{2,4}{\sqrt{64}} = \frac{2,4}{8} = 0,3 \text{ Stunden (18 Min)}$$
* **Sicherheitsfaktor ($z$):** Für ein 95 % Vertrauensniveau nutzen wir den Z-Wert $1,96$.
* **Fehlerspanne:** $1,96 \cdot 0,3 \approx 0,59$

**Das Intervall:**
$$12 \pm 1,96 \cdot 0,3 \implies [11,41; 12,59]$$

---

##### 💡 Fazit für Ingenieure & Marketing
Kein Hersteller kennt das „echte“ $\sigma$ aller Millionen produzierten Akkus. Durch die Schätzung mit $S$ aus der Stichprobe und das Vertrauen in den ZGS wissen wir:

> „Ich bin mir zu **95 % sicher**, dass der wahre Durchschnitt aller Akkus dieser Serie zwischen **11 Stunden 25 Minuten** und **12 Stunden 35 Minuten** liegt.“

Das Marketing hat nun eine verlässliche Basis und kann sicher mit "über 11 Stunden" werben, ohne zu lügen.

```python
import numpy as np
from scipy.stats import norm, t

# 1. Feste Stichprobe von n=64 Smartphones (berechnet für Mean=12, S=2.4)
akku_zeiten = [
    7.94, 8.07, 8.2, 8.33, 8.46, 8.58, 8.71, 8.84, 8.97, 9.1,
    9.23, 9.36, 9.49, 9.62, 9.74, 9.87, 10.0, 10.13, 10.26, 10.39,
    10.52, 10.65, 10.78, 10.9, 11.03, 11.16, 11.29, 11.42, 11.55, 11.68,
    11.81, 11.94, 12.06, 12.19, 12.32, 12.45, 12.58, 12.71, 12.84, 12.97,
    13.1, 13.22, 13.35, 13.48, 13.61, 13.74, 13.87, 14.0, 14.13, 14.26,
    14.38, 14.51, 14.64, 14.77, 14.9, 15.03, 15.16, 15.29, 15.42, 15.54,
    15.67, 15.8, 15.93, 16.06
]

# 2. Kennzahlen aus der festen Stichprobe berechnen
n = len(akku_zeiten)
x_bar = np.mean(akku_zeiten)
# In np.std sorgt ddof=1 für die erwartungstreue Stichprobenstandardabweichung s
s = np.std(akku_zeiten, ddof=1)

# 3. Intervall-Parameter
confidence = 0.95
# Standardfehler SE = S / sqrt(n)
se = s / np.sqrt(n)

# Z-Faktor für 95% (zweiseitig) -> ergibt 1.96
z_factor = norm.ppf(1 - (1 - confidence) / 2)

# Alternative: t-verteilung
# Wenn n = 64, dann ist df = 63
t_factor = t.ppf(1 - (1 - confidence) / 2, df=n-1)
# Die All-in-One Lösung für die t-Verteilung
t_intervall = t.interval(confidence, df=n-1, loc=x_bar, scale=s/np.sqrt(n))

# Fehlerspanne (Margin of Error)
margin_of_error = z_factor * se

# Konfidenzintervall berechnen
untergrenze = x_bar - margin_of_error
obergrenze = x_bar + margin_of_error

# 4. Ausgabe der Ergebnisse
print(f"--- Analyse: Akku-Stresstest (Feste Daten) ---")
print(f"Anzahl Smartphones (n):  {n}")
print(f"Mittelwert (x_bar):      {x_bar:.2f} Stunden")
print(f"Standardabweichung (s):  {s:.2f} Stunden")
print(f"Standardfehler (SE):     {se:.2f} Stunden")
print(f"Z-Faktor (95%):          {z_factor:.2f}")
print(f"-----------------------------------------------")
print(f"95% Konfidenzintervall:  [{untergrenze:.2f}h, {obergrenze:.2f}h]")
print(f"In Minuten:              Zwischen {int(untergrenze)}h {int((untergrenze%1)*60)}min "
      f"und {int(obergrenze)}h {int((obergrenze%1)*60)}min")
print(f"z-Faktor vs t-Faktor: {z_factor} vs. {t_factor}")
print(f"95% t-Verteilng Konfidenzintervall:  [{t_intervall[0]:.2f}h, {t_intervall[1]:.2f}h]")
# All-in-One Lösung Normalverteilung
x_bar = 12.0
s = 2.4
n = 64
confidence = 0.95

# Die All-in-One Zeile
# loc = Mittelwert
# scale = Standardfehler (s / Wurzel n)
untergrenze, obergrenze = norm.interval(confidence,
                                        loc=x_bar,
                                        scale=s/np.sqrt(n))

print(f"95% KI (Normal): [{untergrenze:.2f}, {obergrenze:.2f}]")
```

#### 🔩 Beispiel: Die Schrauben-Fabrik (Konfidenzintervall für die Varianz)

In diesem Beispiel untersuchen wir nicht den Durchschnitt, sondern die **Präzision** einer Maschine. Wir wollen wissen, wie stark der Durchmesser der produzierten Spezialschrauben schwankt.

##### 1. Ausgangslage
Wir ziehen eine Stichprobe von **$n = 10$** Schrauben. Die Messung ergibt eine Stichprobenvarianz von:
* **Stichprobenvarianz ($s^2$):** $0,04 \text{ mm}^2$
* **Sicherheitsniveau:** $99\% \ (\alpha = 0,01)$

Da die Varianz nicht normalverteilt ist, sondern einer **Chi-Quadrat-Verteilung ($\chi^2$)** folgt, benötigen wir die entsprechenden Werte für $df = n - 1 = 9$ Freiheitsgrade.

##### 2. Kritische Werte (Quantile)
Aus der $\chi^2$-Tabelle (oder via Python) entnehmen wir für $99\%$ Vertrauen:
* $\chi^2_{0,995; 9} \approx 23,59$ (oberer Wert)
* $\chi^2_{0,005; 9} \approx 1,73$ (unterer Wert)

##### 3. Berechnung des Intervalls
Die Formel für das Konfidenzintervall der Varianz lautet:
$$\left[ \frac{(n-1) \cdot s^2}{\chi^2_{1-\alpha/2; n-1}} \ ;\ \frac{(n-1) \cdot s^2}{\chi^2_{\alpha/2; n-1}} \right]$$

Einsetzen der Werte:
$$\left[ \frac{9 \cdot 0,04}{23,59} \ ;\ \frac{9 \cdot 0,04}{1,73} \right] = [0,015 \text{ mm}^2 \ ;\ 0,208 \text{ mm}^2]$$

---

##### 💡 Interpretation
Um zu **99 % sicher** zu sein, den wahren Wert der Varianz einzufangen, müssen wir akzeptieren, dass die wahre Varianz im schlimmsten Fall deutlich höher sein könnte als unsere Messung ($0,208$ vs. $0,04$).

Dies zeigt, wie vorsichtig man bei der Schätzung von Streuungen mit kleinen Stichproben sein muss!

```python
from scipy.stats import chi2

# 1. Feste Liste von n=10 Schrauben-Durchmessern (in mm)
schrauben_daten = [
    9.71, 9.80, 9.88, 9.95, 10.01, 10.07, 10.14, 10.22, 10.31, 10.42
]

# 2. Kennzahlen berechnen
n = len(schrauben_daten)
df = n - 1
# ddof=1 für die Schätzfunktion s^2
s2 = np.var(schrauben_daten, ddof=1)

# 3. Parameter für 99% Konfidenzintervall
confidence = 0.99
alpha = 1 - confidence

# Kritische Werte aus der Chi-Quadrat-Verteilung
chi2_lower = chi2.ppf(alpha / 2, df)
chi2_upper = chi2.ppf(1 - alpha / 2, df)

# 4. Intervall berechnen
untergrenze = (df * s2) / chi2_upper
obergrenze = (df * s2) / chi2_lower

# 5. Ausgabe
print(f"--- Analyse: Schrauben-Varianz ---")
print(f"Stichprobengröße (n):   {n}")
print(f"Gemessene Varianz (s2): {s2:.3f} mm^2")
print(f"Sicherheitsniveau:      {confidence*100}%")
print(f"----------------------------------")
print(f"Chi2-Werte:             unten: {chi2_lower:.2f}, oben: {chi2_upper:.2f}")
print(f"99% Konfidenzintervall: [{untergrenze:.3f} mm^2, {obergrenze:.3f} mm^2]")

# Kleiner Bonus: Intervall für die Standardabweichung (einfach Wurzel ziehen)
print(f"Intervall für s:        [{np.sqrt(untergrenze):.3f} mm, {np.sqrt(obergrenze):.3f} mm]")
```

#### 🐿️ Beispiel: Die Nuss-Horter-Studie (Bootstrapping)

Wir untersuchen, wie viele Haselnüsse Eichhörnchen für den Winter verstecken. Wir haben **$n=100$** Eichhörnchen im Stadtpark beobachtet.

##### Die "verrückte" Verteilung
Die Daten sind absolut nicht normalverteilt:
* **Die Mehrheit:** Versteckt zwischen 200 und 500 Nüssen.
* **Die "Profis":** Ein paar wenige sammeln bis zu 3.000 Nüsse (extreme Ausreißer).
* **Die "Verpeilten":** Einige verstecken fast gar nichts.

Bei so einer Verteilung würde der **Mittelwert** durch die Super-Horter völlig nach oben verzerrt werden. Wir wollen daher den **Median** wissen – also was das "typische" Eichhörnchen tut.

##### Warum Bootstrapping?
Es gibt keine einfache Standard-Formel, um ein Konfidenzintervall für den Median einer so schiefen "Eichhörnchen-Verteilung" zu berechnen. Bootstrapping ist hier der einzige Weg: Wir lassen den Computer 10.000 Mal "so tun als ob" und schauen, wie stabil unser Median ist.

---

##### 📊 Die Eckdaten der Stichprobe ($n=100$)
* **Median der Beobachtungen:** $415$ Nüsse
* **Spannweite:** $12$ bis $2.950$ Nüsse

```python
import numpy as np
from scipy.stats import bootstrap

# 1. Unsere Eichhörnchen-Daten (n=100)
eichhoernchen_daten = [
    12, 45, 88, 120, 150, 180, 200, 210, 220, 230, 240, 250, 260, 270, 280,
    290, 300, 310, 320, 330, 340, 350, 360, 370, 380, 390, 400, 405, 410, 415,
    415, 415, 415, 420, 425, 430, 435, 440, 445, 450, 455, 460, 465, 470, 475,
    480, 485, 490, 495, 500, 510, 520, 530, 540, 550, 560, 570, 580, 590, 600,
    620, 640, 660, 680, 700, 720, 740, 760, 780, 800, 850, 900, 950, 1000, 1100,
    1200, 1300, 1400, 1500, 1600, 1700, 1800, 1900, 2000, 2100, 2200, 2300, 2400,
    2500, 2600, 2700, 2800, 2900, 2950, 50, 60, 70, 80, 90, 100
]

# Wichtig: Die Daten müssen als Sequenz in einem Tuple übergeben werden!
data = (eichhoernchen_daten,)

# 2. Die All-in-One Bootstrap Funktion
res = bootstrap(data,
                statistic=np.median,
                confidence_level=0.95,
                n_resamples=10000,
                method='percentile')

# 3. Ergebnisse ausgeben
print(f"Media (Punktschätzer): {np.median(eichhoernchen_daten):.1f}")
print(f"95% KI Untergrenze:    {res.confidence_interval.low:.1f}")
print(f"95% KI Obergrenze:     {res.confidence_interval.high:.1f}")
print(f"Standardfehler:        {res.standard_error:.2f}")
```
