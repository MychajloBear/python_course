# Aufgabe 1: Das Gesetz der großen Zahlen am Computer simulieren (Münze)

- In dieser Aufgabe nutzen wir Python, um das Gesetz der großen Zahlen zu visualisieren. Anstatt hunderte Male selbst eine Münze zu werfen, lassen wir den Computer die Münze 1.000-mal werfen und berechnen laufend die **relative Häufigkeit** für das Ergebnis „Kopf“.
- Visualisiere anschließend deine Ergebnisse in einem Diagramm.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameter festlegen
np.random.seed(42)
n_versuche = 1000
wahrscheinlichkeit_kopf = 0.5  # theoretische Wahrscheinlichkeit für Kopf

# 2. Münzwürfe simulieren (0 = Zahl, 1 = Kopf)
wuerfe = np.random.randint(0, 2, size=n_versuche)
print("Erste 20 Würfe (0 = Zahl, 1 = Kopf):", wuerfe[:20])

# 3. Relative Häufigkeit für Kopf (1) berechnen
absolute_haeufigkeiten = np.cumsum(wuerfe)
relative_haeufigkeiten = absolute_haeufigkeiten / np.arange(1, n_versuche + 1)
print(f"Finale relative Häufigkeit nach {n_versuche} Würfen: {relative_haeufigkeiten[-1]:.4f}")

# 4. Visualisierung
plt.figure(figsize=(10, 6))
plt.plot(np.arange(1, n_versuche + 1), relative_haeufigkeiten, label='Relative Häufigkeit Kopf', color='steelblue')
plt.axhline(y=wahrscheinlichkeit_kopf, color='red', linestyle='--', label='Theoretische Wahrscheinlichkeit (0.5)')
plt.title('Gesetz der großen Zahlen: Münzwurf-Simulation (1.000 Versuche)')
plt.xlabel('Anzahl der Würfe')
plt.ylabel('Relative Häufigkeit')
plt.ylim(0, 1)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

# Aufgabe 2: Das Gesetz der großen Zahlen am Computer simulieren (Würfel)

- In dieser Aufgabe nutzen wir Python, um das Gesetz der großen Zahlen zu visualisieren. Anstatt hunderte Male selbst zu würfeln, lassen wir den Computer in Sekundenschnelle 1.000 Würfe durchführen und berechnen laufend den Durchschnitt der Augenzahlen.
- Visualisiere anschließend deine Ergebnisse in einem Diagramm.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameter festlegen
np.random.seed(42)
n_wuerfe = 1000
erwartungswert_wuerfel = 3.5  # theoretischer Durchschnitt eines fairen 6-seitigen Würfels (1+2+3+4+5+6)/6

# 2. Würfelwürfe simulieren (Zahlen von 1 bis 6)
wuerfel_wuerfe = np.random.randint(1, 7, size=n_wuerfe)
print("Erste 20 Würfelwürfe:", wuerfel_wuerfe[:20])

# 3. Laufenden Durchschnitt (kumulativen Mittelwert) berechnen
kumulative_summe = np.cumsum(wuerfel_wuerfe)
laufender_durchschnitt = kumulative_summe / np.arange(1, n_wuerfe + 1)
print(f"Finaler Durchschnitt nach {n_wuerfe} Würfen: {laufender_durchschnitt[-1]:.4f}")

# 4. Visualisierung
plt.figure(figsize=(10, 6))
plt.plot(np.arange(1, n_wuerfe + 1), laufender_durchschnitt, label='Laufender Durchschnitt der Augenzahlen', color='darkgreen')
plt.axhline(y=erwartungswert_wuerfel, color='red', linestyle='--', label='Theoretischer Erwartungswert (3.5)')
plt.title('Gesetz der großen Zahlen: Würfel-Simulation (1.000 Versuche)')
plt.xlabel('Anzahl der Würfe')
plt.ylabel('Durchschnittliche Augenzahl')
plt.ylim(1, 6)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
