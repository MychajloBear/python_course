# Aufgabe 1: Das Gesetz der großen Zahlen am Computer simulieren (Münze)

- In dieser Aufgabe nutzen wir Python, um das Gesetz der großen Zahlen zu visualisieren. Anstatt hunderte Male selbst eine Münze zu werfen, lassen wir den Computer die Münze 1.000-mal werfen und berechnen laufend die **relative Häufigkeit** für das Ergebnis „Kopf“.
- Visualisiere anschließend deine Ergebnisse in einem Diagramm.


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameter festlegen
n_versuche = 100
wahrscheinlichkeit_kopf = 0.5  # Theoretische Wahrscheinlichkeit für Kopf

# 2. Münzwürfe simulieren (0 = Zahl, 1 = Kopf)
# random.randint(0, 2) erzeugt zufällig 0 oder 1
wuerfe = np.random.randint(0, 2, size=n_versuche)

# 3. Laufende relative Häufigkeit für "Kopf" (1) berechnen
# Da Kopf = 1 und Zahl = 0 ist, entspricht die Summe der Würfe der Anzahl an "Kopf"
relative_haeufigkeiten = np.cumsum(wuerfe) / np.arange(1, n_versuche + 1)

# 4. Plot erstellen
plt.figure(figsize=(10, 6))
plt.plot(relative_haeufigkeiten, color='darkorange', label='Relative Häufigkeit "Kopf"', lw=2)

# Horizontale Linie für die theoretische Wahrscheinlichkeit (0.5)
plt.axhline(y=wahrscheinlichkeit_kopf, color='black', linestyle='--', label=f'Theoretische Wahrscheinlichkeit p = {wahrscheinlichkeit_kopf}')

# Styling
plt.title('Gesetz der großen Zahlen: Münzwurf-Simulation')
plt.xlabel('Anzahl der Würfe (n)')
plt.ylabel('Relative Häufigkeit')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

# Aufgabe 2: Das Gesetz der großen Zahlen am Computer simulieren (Würfel)

- In dieser Aufgabe nutzen wir Python, um das Gesetz der großen Zahlen zu visualisieren. Anstatt hunderte Male selbst zu würfeln, lassen wir den Computer in Sekundenschnelle 1.000 Würfe durchführen und berechnen laufend den Durchschnitt der Augenzahlen.
- Visualisiere anschließend deine Ergebnisse in einem Diagramm.
-


```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameter festlegen
n_versuche = 1000
erwartungswert = 3.5  # (1+2+3+4+5+6) / 6

# 2. Würfelwürfe simulieren (Zahlen von 1 bis 6)
würfe = np.random.randint(1, 7, size=n_versuche)

# 3. Laufenden Durchschnitt berechnen
# cumsum addiert die Werte schrittweise auf, arange teilt durch die Anzahl
durchschnitte = np.cumsum(würfe) / np.arange(1, n_versuche + 1)

# 4. Plot erstellen
plt.figure(figsize=(10, 6))
plt.plot(durchschnitte, color='C0', label='Laufender Durchschnitt', lw=2)

# Horizontale Linie für den wahren Erwartungswert (3,5)
plt.axhline(y=erwartungswert, color='red', linestyle='--', label=f'Erwartungswert μ = {erwartungswert}')

# Styling
plt.title('Gesetz der großen Zahlen: Konvergenz beim Würfeln')
plt.xlabel('Anzahl der Würfe (n)')
plt.ylabel('Durchschnittswert')
plt.ylim(1, 6)  # Bereich der Würfelseiten
plt.legend()
plt.grid(alpha=0.3)
```
