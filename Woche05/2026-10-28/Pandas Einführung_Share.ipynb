# Grundlagen von Pandas

Pandas ist eine beliebte Open-Source-Programmbibliothek, die speziell für die Datenanalyse und Datenmanipulation entwickelt wurde. Man kann sich Pandas wie eine mächtige, programmierbare Version von Excel vorstellen, die auch sehr große Datenmengen extrem schnell verarbeiten kann.

## 1. Das Pandas Dataframe
Pandas bietet verschiedene vorgefertigte Datenstrukturen, um uns das Arbeiten mit Daten unterschiedlichster Struktur so einfach wie möglich zu machen. Abhängig von der Struktur der Daten gibt es hierfür entweder die sogenannten *Series* oder die *Dataframes*.

Eine Pandas Series ist eine eindimensionale, beschriftete Liste, deren Felder quasi jeden Datentyp (Int, Float, String, Python Objects,...) annehmen kann. Sie ist vergleichbar mit einer einzelnen Spalte in einer Tabelle. Mithilfe von Indizes können wir sehr leicht auf die einzelnen Datenfelder in der Series zugreifen.

Für unsere Zwecke nutzen wir jedoch zu 99% das Pandas-Dataframe. Es ist quasi die zweidimensionale Version einer Series und erlaubt uns, jegliche Datentypen in einer zweidimensionalen, tabellarischen Datenstruktur unterzubringen, mit der sich sehr komfortabel weiterarbeiten lässt.

## 2. Daten laden

Erstmal wollen wir die **CSV-Dateien als Pandas-DataFrames** importieren.
- market_data
- sales_data

```python
# Pandas Bibliothek importieren
import pandas as pd

# Dateien importieren
market_data = pd.read_csv('project_data/market_data.csv')
sales_data = pd.read_csv('project_data/sales_data.csv')
```


## 3. Untersuchung der DataFrame-Struktur

Nach dem Laden mehrerer **CSV-Dateien als Pandas-DataFrames**, sollen folgende **Strukturelemente** untersucht werden:  
- **Spaltennamen**  
- **Datentypen**  
- **Anzahl der Zeilen und Spalten**  


```python
# Überprüfung der geladenen DataFrames
print(market_data.describe())
print("###############")
market_data.info()
print("###############")
market_data.head(10)
```

Wir haben nun also einen ersten Eindruck von den Daten und ihrer Struktur erhalten - offenbar handelt es sich um Inventar-Daten eines Supermarktes oder anderen Geschäfts, das Obst und Gemüse verkauft.
Was aber, wenn wir einen genaueren Blick in die Daten werfen möchten? Hierfür bietet die Pandas-Bibliothek mehrere Möglichkeiten:


Wenn wir nur eine spezifische Spalte auswählen möchten, können wir wie bei Python-Listen eckige Klammern und einen Index benutzen. praktisch ist hierbei, dass wir den Namen einer Spalte als Index benutzen können.

```python
print(market_data['Produkt'])
```

Wir können mit ebenjener Syntax auch Listen als Argument übergeben, um mehrere Spalten auszuwählen:

```python
print(market_data[['Produkt', 'Verkaufte Menge']])
```

Möchten wir stattdessen auf spezifische *Zeilen* zugreifen, können wir `loc[]` oder `iloc[]` benutzen:

```python
print(market_data.iloc[0], "\n")

print(market_data.loc[market_data['Produkt'] == 'Zitrone'])
```

Zuletzt stellt sich noch die Frage: was, wenn ich *Bereiche* statt einzelner Spalten/Zeilen auswählen möchte?
Wir erinnern uns (hoffentlich) an Python Listen, und wie man es dort umgesetzt hat.

```python
#print(market_data.iloc[1:20:1, 0:4:1])
# [ZeileStart:ZeileStop,ZeileStep, SpalteStart:SpalteStop:SpalteStep]
display(market_data.loc[1:5, "Produkt":"Preis pro Einheit (EUR)"])
```

## 4. Weitere Dataframe Funktionen

Wir können die Syntax für Indexing auch nutzen, um nur Felder zu erhalten, die eine bestimmte Kondition erfüllen

```python
filtered_market = market_data[market_data['Preis pro Einheit (EUR)'] < 0.50]
print(filtered_market) # Printed die Daten als Text (trotzdem in Tabellenform)
display(filtered_market) # Stellt die Daten als Tabelle dar
```

Wir können ebenfalls Sortierungen durchführen, um unsere Dataframes in eine übersichtlichere Form zu bringen.

```python
sorted_market = market_data.sort_values(by="Preis pro Einheit (EUR)", ascending=False)
display(sorted_market.head(10))
```

Besonders hilfreich ist auch die Möglichkeit, gesamte (numerische) Spalten miteinander zu verrechnen. Wir können z.B. Spalten Wert-für-Wert miteinander multiplizieren und ohne Zwischenschritt eine neue Spalte it den Ergebnissen erzeugen!


```python
market_data['Gesamtwert'] = market_data['Preis pro Einheit (EUR)'] * market_data['Verkaufte Menge']
print(market_data.head())
```


Als nächstes wollen wir die Produkte finden, deren Verfallsdatum bereits überschritten wurde. Wie können wir das umsetzen?

```python
abgelaufen = market_data[market_data['Verfallsdatum'] < "2025-01-15"]
abgelaufen.sort_values(by="Verfallsdatum")
```

Wir können auch Felder mit spezifischen Werten filtern.


```python
apfel_banane = market_data[(market_data["Produkt"] == "Apfel") | (market_data["Produkt"] == "Banane")]
apfel_banane.head(10)
```
