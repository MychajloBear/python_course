## Kapitel 1 – Die Daten treffen ein


Die Verkaufsdaten wurden als CSV-Datei exportiert und liegen im Projektordner.

1. Lies die Datei `schokoladenhandel.csv` in ein DataFrame `df` ein.
2. Gib anschließend die ersten fünf Zeilen aus.
3. Gib danach die letzten fünf Zeilen aus.

**Hinweis:** Nutze `pd.read_csv()`, `head()` und `tail()`.


```python
import pandas as pd

df = pd.read_csv("schokoladenhandel.csv")

print(df.head())
print(df.tail())
```

```python
print("--- INFO ---")
df.info()

print("\n--- DESCRIBE ---")
print(df.describe())
```

## Kapitel 3 – Die Geschäftsleitung fragt nach


Die Geschäftsführung möchte wissen, wie umfangreich die Verkaufsdaten sind.

Bearbeite folgende Aufgaben:

- Gib alle Spaltennamen aus.

Versuche möglichst ohne manuelles Zählen auszukommen.


```python
print(f"Zeilen und Spalten: {df.shape}")

print("\nSpaltennamen:")
print(df.columns)

print("\nDatentypen:")
print(df.dtypes)
```

## Kapitel 4 – Einzelne Bestellungen prüfen


Vor einer Besprechung sollen einzelne Bestellungen kontrolliert werden.

1. Gib mit `iloc` die dritte Datenzeile aus.
2. Wähle mit `loc` die Zeilen 5 bis 10 sowie die Spalten `Produkt` und `Umsatz`.
3. Gib ausschließlich den Umsatz der ersten Bestellung aus.

Achte darauf, wann Positionen und wann Bezeichnungen verwendet werden.


```python
print(df.iloc[2],"\n")

print(df.loc[5:10, ["Produkt", "Umsatz"]],"\n")

print(df.loc[0, "Umsatz"])
```

## Kapitel 5 – Bericht für die Region Nord


Die Regionalleitung Nord möchte nur ihre eigenen Verkaufszahlen sehen.

1. Erstelle ein neues DataFrame mit allen Bestellungen der Region `Nord`.
2. Bestimme, wie viele Bestellungen darin enthalten sind.
3. Zeige die ersten fünf Zeilen dieses DataFrames an.

Überlege, weshalb das Original-DataFrame dabei unverändert bleiben sollte.


```python
df_nord = df[df["Region"] == "Nord"].copy()

print(f"Bestellungen in Nord: {len(df_nord)}")

print("\nErsten 5 Zeilen:")
print(df_nord.head())
```

## Kapitel 6 – Die erfolgreichsten Verkäufe


Die Geschäftsführung interessiert sich für besonders erfolgreiche Verkäufe.

1. Sortiere das DataFrame nach `Umsatz` in absteigender Reihenfolge.
2. Gib die fünf umsatzstärksten Bestellungen aus.
3. Welche Produkte erscheinen besonders häufig unter den Top 5?


```python
df_sorted = df.sort_values(by="Umsatz", ascending=False)

print("Top 5:")
print(df_sorted.head(5),"\n")

print("Häufigste Produkte (Top 5):")
print(df_sorted.head(5)["Produkt"].value_counts(),"\n")
```

## Kapitel 7 – Marketing plant eine Rabattaktion


Für eine Werbekampagne sollen alle Produkte mit 10 % Rabatt angeboten werden.

1. Erstelle die Spalte `Rabattpreis`.
2. Berechne daraus eine neue Spalte `Rabattumsatz`.
3. Vergleiche die ersten fünf Zeilen mit den ursprünglichen Preisen.

Nutze mathematische Operationen direkt auf ganzen DataFrame-Spalten.


```python
df["Rabattpreis"] = df["Preis_pro_Packung"] * 0.9

df["Rabattumsatz"] = df["Verkaufte_Packungen"] * df["Rabattpreis"]

print("Vergleich:")
print(df[["Produkt", "Preis_pro_Packung", "Rabattpreis", "Umsatz", "Rabattumsatz"]].head())
```

## Kapitel 8 – Bericht für die Präsentation


Für eine Präsentation werden nur wenige Informationen benötigt.

1. Erstelle mit Slicing ein DataFrame mit den ersten zehn Datensätzen.
2. Wähle daraus ausschließlich die Spalten
   - `Produkt`
   - `Region`
   - `Verkaufte_Packungen`
   - `Umsatz`

Das Ergebnis soll möglichst übersichtlich sein.


```python
df_presentation = df[["Produkt", "Region", "Verkaufte_Packungen", "Umsatz"]].iloc[:10]

print("Übersicht für die Präsentation:")
print(df_presentation)
```

## Kapitel 9 – Große Bestellungen


Der Geschäftsführer interessiert sich nur für größere Bestellungen.

1. Filtere alle Bestellungen mit mehr als 100 verkauften Packungen.
2. Sortiere das Ergebnis nach `Verkaufte_Packungen` absteigend.
3. Gib anschließend nur die Spalten
   - `Produkt`
   - `Region`
   - `Verkaufte_Packungen`
   - `Umsatz`
aus.


```python
df_big = df[df["Verkaufte_Packungen"] > 100].sort_values(by="Verkaufte_Packungen", ascending=False)

print(df_big[["Produkt", "Region", "Verkaufte_Packungen", "Umsatz"]])
```

Zum Abschluss soll ein kleiner Managementbericht vorbereitet werden.

Erstelle ein neues DataFrame, das folgende Bedingungen erfüllt:

Gib nur die Spalten

- `Rabattpreis`
- `Rabattumsatz`

aus.

Diese Aufgabe kombiniert mehrere der im Praktikum behandelten Konzepte.


```python
df_final = df[(df["Region"] == "Nord") & (df["Verkaufte_Packungen"] > 100)].copy()

df_final_sorted = df_final.sort_values(by="Rabattumsatz", ascending=False)

print(df_final[["Produkt", "Verkaufte_Packungen", "Rabattpreis", "Rabattumsatz"]])
```
